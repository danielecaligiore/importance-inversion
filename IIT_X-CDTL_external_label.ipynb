```python
# ==============================================================================
# === X-CDTL Framework | Module 1: Topological Characterization & IIT Logic ===
# ==============================================================================
# Researcher: Daniele Caligiore
# Affiliation: CNR - Italian National Research Council (ISTC)
# Version: 4.0 (Revised September 2026 - Nature Communications Submission)
# ------------------------------------------------------------------------------
# DESCRIPTION:
# This module characterizes graph ensembles from four heterogeneous domains
# (Social, Molecular, Protein, and Linguistic) using a common set of
# multi-scale topological descriptors. It generates domain-specific functional
# labels or controlled perturbation labels for downstream validation, while
# keeping these labels separate from anchor discovery.
#
# The module ranks topological descriptors according to their domain-
# discriminative utility, evaluates pairwise structural compatibility in a
# common standardized space, and identifies candidate structural anchors
# through the Importance Inversion Transfer (IIT) score. Anchor discovery uses
# training observations only; held-out observations remain outside feature
# importance estimation and pairwise IIT calculation.
# ------------------------------------------------------------------------------
# DUAL-PARADIGM CONFIGURATION:
# The validation regime is selected through the global variable
# 'VALIDATION_PARADIGM':
# 1. "functional_validation": Uses domain-specific functional labels or
#    controlled perturbation labels for downstream validation.
# 2. "structural_verification": Retains the original circular benchmark based
#    on multivariate topological extremeness (90th-percentile z-score sum).
# ------------------------------------------------------------------------------
# CORE PIPELINE:
# 1. Graph construction and multi-scale topological feature extraction.
# 2. Inductive domain classification and consensus Borda ranking of the
#    twelve topological descriptors using training observations only.
# 3. Pairwise IIT scoring based on standardized mean proximity and mean-
#    difference alignment, followed by selection of eight candidate structural
#    anchors.
# 4. Global consensus aggregation of pairwise IIT scores to derive a general
#    candidate structural-anchor set for downstream transfer experiments.

import sys, os, subprocess

# ------------------------------------------------------------
# === ENVIRONMENT CHECK ===
# ------------------------------------------------------------
expected_venv = "ml-env"
if expected_venv in sys.prefix:
    print(f"✅ Kernel uses the correct virtual environment: {expected_venv}")
else:
    print(f"⚠️ Warning: current kernel not in {expected_venv} → please select ml-env/bin/python")

# ------------------------------------------------------------
# === GPU DETECTION & ASSIGNMENT (MUST BE BEFORE TORCH IMPORT) ===
# ------------------------------------------------------------
def find_free_gpu():
    """Return index of GPU with most free memory."""
    try:
        smi_output = subprocess.check_output(
            "nvidia-smi --query-gpu=index,memory.free --format=csv,noheader,nounits",
            shell=True
        ).decode("utf-8").strip().split("\n")
        gpus = [(int(line.split(",")[0]), int(line.split(",")[1])) for line in smi_output]
        gpus.sort(key=lambda x: x[1], reverse=True)
        return str(gpus[0][0])
    except Exception as e:
        print("⚠️ Unable to detect free GPU:", e)
        return "0"

# Set GPU visibility before importing PyTorch so that the selected device
# is recognized during CUDA initialization.
os.environ["CUDA_VISIBLE_DEVICES"] = find_free_gpu()
print(f"🌍 OS level: Target GPU set to index {os.environ['CUDA_VISIBLE_DEVICES']}")

# ------------------------------------------------------------
import torch

if torch.cuda.is_available():
    current_device = torch.cuda.current_device()
    gpu_name = torch.cuda.get_device_name(current_device)
    free_mem, total_mem = torch.cuda.mem_get_info()
    
    print(f"🖥️ PyTorch correctly initialized on GPU: {gpu_name}")
    print(f"📊 VRAM Status: {free_mem//1024**2} MB free / {total_mem//1024**2} MB total")
    device = torch.device("cuda")
else:
    print("⚠️ GPU not available, using CPU")
    device = torch.device("cpu")

# ------------------------------------------------------------
# === IMPORTS ===
# ------------------------------------------------------------
import random, warnings
warnings.filterwarnings("ignore")

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

from torch_geometric.datasets import QM9, TUDataset
import networkx as nx, pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
from sklearn.model_selection import StratifiedKFold, train_test_split
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, confusion_matrix
from community import community_louvain
import nltk
nltk.download('brown', quiet=True)
from nltk.corpus import brown
from itertools import permutations

# ------------------------------------------------------------
# === VALIDATION STRATEGY CONFIGURATION ===
# ------------------------------------------------------------
# "structural_verification" (Original): Anomalies defined via topological z-scores.
# "functional_validation"   (Updated): Anomalies defined via exogenous/functional labels.
VALIDATION_PARADIGM = "functional_validation"
#VALIDATION_PARADIGM = "structural_verification"

print(f"Active Framework Mode: {VALIDATION_PARADIGM.upper()}")

# ------------------------------------------------------------
# === GLOBAL SEED CONTROL ===
# ------------------------------------------------------------
target_seeds = [1, 24, 42, 50, 123, 501, 700, 800, 920, 999]

# ------------------------------------------------------------
# === INITIAL REPRODUCIBILITY SETUP (MASTER SEED) ===
# ------------------------------------------------------------
# CRITICAL: To ensure that graph pools and perturbations are identical 
# across all executions, we establish a MASTER_POOL_SEED before any
# data processing or sampling occurs.
MASTER_POOL_SEED = 24

def set_global_seed(seed=24):
    """Encapsulates all RNG resets for full determinism."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

# Initialize the environment with the master seed immediately
set_global_seed(MASTER_POOL_SEED)
# ------------------------------------------------------------
# === GRAPH METRIC COMPUTATION ===
# ------------------------------------------------------------
def largest_component(g):
    """Return the largest connected component of a graph."""
    return g if nx.is_connected(g) else g.subgraph(max(nx.connected_components(g), key=len)).copy()

def compute_metrics(g):
    """
    Compute multi-scale structural metrics. 
    n_nodes represents the total physical size (raw), 
    while topological metrics are calculated on the Largest Connected Component.
    """
    # 1. RAW SIZE: Total nodes in the graph as constructed.
    # Raw node count is retained so that edge-removal and word-order perturbations
    # do not introduce a node-count change in the corresponding controlled
    # comparisons.
    n_raw = g.number_of_nodes()
    e_raw = g.number_of_edges()
    
    # 2. Extract LCC for topological descriptors
    # We must use the LCC because diameter/shortest_path are undefined for disconnected graphs.
    gc = largest_component(g)
    n_lcc = gc.number_of_nodes()
    
    # Safeguard: if the graph is too fragmented to be analyzed
    if n_lcc < 2:
        return {m: np.nan for m in metrics}
    
    # Calculate all descriptors on the LCC
    A = nx.to_numpy_array(gc)
    eig_A = np.linalg.eigvals(A)
    spectral_radius = float(np.max(np.abs(np.real(eig_A))))
    L = nx.normalized_laplacian_matrix(gc).toarray()
    eig_L = np.real(np.linalg.eigvals(L))
    lambda_2 = sorted(eig_L)[1] if len(eig_L) > 1 else 0
    try:
        asp = nx.average_shortest_path_length(gc)
        diam = nx.diameter(gc)
    except Exception:
        asp, diam = np.nan, np.nan
    partition = community_louvain.best_partition(
        gc,
        random_state=MASTER_POOL_SEED
    )
    modularity = community_louvain.modularity(
        partition,
        gc
    )
    
    return dict(
        n_nodes=n_raw,           # Uses raw size (no size bias)
        n_edges=e_raw,           # Raw edge count
        density=nx.density(g),   # Calculated on the full graph
        avg_clustering=nx.average_clustering(gc),
        transitivity=nx.transitivity(gc),
        assortativity=nx.degree_assortativity_coefficient(gc),
        efficiency=nx.global_efficiency(gc),
        avg_shortest_path=asp,
        diameter=diam,
        spectral_radius=spectral_radius,
        lambda_2=lambda_2,
        modularity=modularity
    )

def attach_paradigms(df, metric_list):
    """Attach the structural-verification labels to the graph pool."""
    # Structural-verification benchmark: multivariate topological extremeness
    # defined by the sum of absolute feature-wise z-scores.
    z_scores = (df[metric_list] - df[metric_list].mean()) / (df[metric_list].std() + 1e-9)
    z_sum = z_scores.abs().sum(axis=1)
    df["y_structural"] = (z_sum > np.percentile(z_sum, 90)).astype(int)
    return df

metrics = ["n_nodes","n_edges","density","avg_clustering","transitivity",
           "assortativity","efficiency","avg_shortest_path","diameter",
           "spectral_radius","lambda_2","modularity"]

# ------------------------------------------------------------
# === FIGURE EXPORT UTILITY (journal-quality) ===
# ------------------------------------------------------------
output_dir = "figures_nature"
os.makedirs(output_dir, exist_ok=True)

def save_highres(filename_base, fig=None):
    """Save journal-compatible vector artwork."""
    if fig is None:
        fig = plt.gcf()

    fig.savefig(
        os.path.join(output_dir, f"{filename_base}.pdf"),
        bbox_inches="tight",
        pad_inches=0.02
    )

    fig.savefig(
        os.path.join(output_dir, f"{filename_base}.svg"),
        bbox_inches="tight",
        pad_inches=0.02
    )

    # Raster preview
    fig.savefig(
        os.path.join(output_dir, f"{filename_base}.png"),
        dpi=600,
        bbox_inches="tight",
        pad_inches=0.02
    )

    print(f"📸 Figure saved: {filename_base}.pdf / .svg / .png")

# ------------------------------------------------------------
# === NATURE COMMUNICATIONS FIGURE STYLE ===
# ------------------------------------------------------------
sns.set_theme(style="white")

plt.rcParams.update({
    "font.family": "sans-serif",
    "font.size": 8,
    "axes.labelsize": 8,
    "axes.titlesize": 8,
    "axes.labelweight": "normal",
    "xtick.labelsize": 7,
    "ytick.labelsize": 7,
    "legend.fontsize": 7,
    "axes.linewidth": 0.8,
    "xtick.major.width": 0.8,
    "ytick.major.width": 0.8,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
    "figure.dpi": 150
})

DOMAIN_COLORS = {
    "Social": "#0072B2",
    "Molecular": "#CC79A7",
    "Proteins": "#E69F00",
    "Linguistic": "#56B4E9"
}

def plot_example_graphs_composite(graphs_info, filename=None):
    """
    Create one 2x2 composite figure with four representative domain graphs.
    """
    fig, axes = plt.subplots(
        2,
        2,
        figsize=(6.8, 6.8),
        constrained_layout=True
    )

    axes = axes.flatten()

    for i, (domain, graph, node_color) in enumerate(graphs_info):

        ax = axes[i]

        # Panel label required by Nature
        ax.text(
            -0.08,
            1.05,
            chr(97 + i),
            transform=ax.transAxes,
            fontsize=10,
            fontweight="bold",
            va="top",
            ha="right"
        )

        pos = nx.spring_layout(
            graph,
            k=0.6,
            seed=24
        )
        #pos = nx.kamada_kawai_layout(graph)

        nx.draw_networkx_edges(
            graph,
            pos,
            ax=ax,
            alpha=0.40,
            width=0.8,
            edge_color="#666666"
        )

        nx.draw_networkx_nodes(
            graph,
            pos,
            ax=ax,
            node_size=80,
            node_color=node_color,
            edgecolors="white",
            linewidths=0.6,
            alpha=1.0
        )

        ax.set_title(
            domain,
            fontsize=8,
            fontweight="normal",
            pad=4
        )

        ax.axis("off")
        ax.margins(0.05)

    if filename:
        save_highres(filename, fig=fig)

    return fig

# ------------------------------------------------------------
# === LOAD FOUR GRAPH DOMAINS (POOLING FOR EFFICIENCY) ===
# ------------------------------------------------------------
print("\n📥 Pre-calculating graph pools for 10-seed efficiency...")

# --- Social Domain ---
# A fixed subsampling strategy reduces the number of potentially overlapping
# ego-networks while keeping pool construction reproducible.
# Select every second eligible ego-centre to reduce the number of sampled
# ego-networks and potential overlap among local subgraphs.

url = "https://snap.stanford.edu/data/facebook_combined.txt.gz"
if not os.path.exists("facebook_combined.txt"):
    !wget -q -O facebook_combined.txt.gz {url} && gunzip -f facebook_combined.txt.gz

# Load the full SNAP Facebook dataset
G_fb_full = nx.read_edgelist("facebook_combined.txt", nodetype=int)

# Identify nodes whose ego-networks fit our 10-30 nodes size constraint
eligible_nodes = [n for n in G_fb_full.nodes() if 10 <= nx.ego_graph(G_fb_full, n).number_of_nodes() <= 30]

# Fixed subsampling of eligible ego-centres to reduce the number of
# potentially overlapping ego-networks while keeping the sampling
# procedure reproducible.
spaced_candidates = eligible_nodes[::2]

# Randomize the candidate order with a fixed seed.
random.seed(MASTER_POOL_SEED)
random.shuffle(spaced_candidates)

def get_perturbed_social_pool(limit=1000, anomaly_rate=0.1):
    """
    Creates a social pool where anomalies are generated by collapsing 
    local triadic cohesion, simulating fragmented social circles.
    Uses spaced_candidates to reduce potential dependence caused by overlapping ego-networks.
    """
    graphs = []
    y_func = []
    count = 0
    # Iterate over the subsampled candidate set.
    for node in spaced_candidates:
        G_ego = nx.ego_graph(G_fb_full, node)
        
        # Systematically mark 10% of samples for functional perturbation
        is_anomaly = 1 if (count % int(1/anomaly_rate) == 0) else 0
        
        if is_anomaly:
            # CONTROLLED PERTURBATION: local structural fragmentation.
            ego_node = node
            triadic_edges = [e for e in G_ego.edges() if ego_node not in e]
            
            if len(triadic_edges) > 0:
                # SEEDING: Local seed ensures instance-level reproducibility
                random.seed(count + MASTER_POOL_SEED)
                n_to_remove = int(len(triadic_edges) * 0.7)
                edges_to_remove = random.sample(triadic_edges, n_to_remove)
                G_ego.remove_edges_from(edges_to_remove)
        
        # Re-verify size constraint after potential fragmentation
        if 10 <= G_ego.number_of_nodes() <= 30:
            graphs.append(G_ego)
            y_func.append(is_anomaly)
            count += 1
            
        if len(graphs) == limit: 
            break
        
    df_metrics = pd.DataFrame([compute_metrics(g) for g in graphs])
    df_metrics["y_functional"] = y_func
    return df_metrics, graphs

print("   • Generating Social Pool (Decimated Sampling to prevent Leakage)...")
pool_ego_metrics, pool_ego_graphs = get_perturbed_social_pool(limit=1000, anomaly_rate=0.1)

# --- Molecular Domain (Exogenous Label: HOMO-LUMO Gap) ---
dataset_qm9 = QM9(root="qm9_data")

def pyg_to_nx(data):
    """Converts a PyTorch Geometric data object to a NetworkX graph."""
    G = nx.Graph()
    G.add_nodes_from(range(data.num_nodes))
    G.add_edges_from(data.edge_index.t().tolist())
    return G

# Storage for graphs and their associated chemical properties
pool_mol_graphs = []
mol_external_gap = []

# Extract up to 1000 graphs together with the HOMO-LUMO gap (index 4 in QM9),
# which provides the functional criterion used for molecular validation.
# Lower values correspond to the selected low-gap condition.
for i in range(3000): 
    data = dataset_qm9[i]
    if 10 <= data.num_nodes <= 30:
        pool_mol_graphs.append(pyg_to_nx(data))
        # Store the HOMO-LUMO gap for functional labeling.
        mol_external_gap.append(float(data.y[0, 4]))
    if len(pool_mol_graphs) == 1000:
        break

# Structural feature extraction (X)
pool_mol_metrics = pd.DataFrame([compute_metrics(g) for g in pool_mol_graphs])

# Functional Label (y_functional): 
# Defined by chemical instability (bottom 10% of the energy gap distribution).
gap_threshold = np.percentile(mol_external_gap, 10) 
pool_mol_metrics["y_functional"] = [1 if g < gap_threshold else 0 for g in mol_external_gap]

# --- Protein Domain (Natural Distribution for Dual-Paradigm Support) ---
dataset_prot = TUDataset(root="data_prot", name="PROTEINS")
pool_prot_graphs = []
prot_y_functional = []

print("   • Extracting PROTEINS (10-30 nodes)...")
for data in dataset_prot:
    G_p = nx.Graph()
    G_p.add_nodes_from(range(data.num_nodes))
    G_p.add_edges_from(data.edge_index.numpy().T)
    if 10 <= G_p.number_of_nodes() <= 30:
        pool_prot_graphs.append(G_p)
        # Functional label: enzymes are assigned 0 and non-enzymes are assigned 1.
        # The non-enzyme class is treated as the anomaly class for validation.
        prot_y_functional.append(0 if int(data.y) == 1 else 1)
    if len(pool_prot_graphs) == 1000: break

pool_prot_metrics = pd.DataFrame([compute_metrics(g) for g in pool_prot_graphs])
pool_prot_metrics["y_functional"] = prot_y_functional
# Note: y_structural will be added by the attach_paradigms function later

# --- Linguistic Domain ---
# Normative instances consist of natural English sentences from the News
# genre; anomalous instances preserve the same words but apply a randomized
# word order. The perturbation therefore keeps the number of lexical tokens
# unchanged while altering the resulting word-adjacency topology.

def get_perturbed_lang_pool(limit=1000, anomaly_rate=0.1):
    graphs = []
    y_func = []
    category_sents = list(brown.sents(categories='news'))
    WINDOW = 2
    
    count = 0
    for sent in category_sents:
        words = [w.lower() for w in sent if w.isalpha()]
        if not (10 <= len(words) <= 30): continue
        
        # Assign each instance to the normative or perturbed condition.
        is_anomaly = 1 if (count % int(1/anomaly_rate) == 0) else 0
        
        if is_anomaly:
            # CONTROLLED PERTURBATION: random word-order scrambling.
            # SEEDING: Local seed ensures the shuffle is deterministic for this sentence
            random.seed(count + MASTER_POOL_SEED)
            random.shuffle(words)
        
        G_l = nx.Graph()
        for i in range(len(words) - WINDOW):
            for j in range(1, WINDOW + 1):
                G_l.add_edge(words[i], words[i+j])
        
        if 10 <= G_l.number_of_nodes() <= 30:
            graphs.append(G_l)
            y_func.append(is_anomaly)
            count += 1
            
        if len(graphs) == limit: break
    
    df_metrics = pd.DataFrame([compute_metrics(g) for g in graphs])
    df_metrics["y_functional"] = y_func
    return df_metrics, graphs

print("   • Generating Linguistic Pool (Natural vs. Scrambled Syntax)...")
pool_lang_metrics, pool_lang_graphs = get_perturbed_lang_pool(limit=1000, anomaly_rate=0.1)

# Attach structural paradigms
pool_lang_metrics = attach_paradigms(pool_lang_metrics, metrics)

# ------------------------------------------------------------------------------
# === PARALLEL VALIDATION LABELS ===
# ------------------------------------------------------------------------------
# Each graph pool retains two labeling schemes:
# 1. y_structural: multivariate topological-extremeness labels used for
#    structural verification.
# 2. y_functional: domain-specific functional or controlled-perturbation labels
#    used for primary functional validation.

print("\n🔄 Integrating Structural (Old) and Functional (New) paradigms...")
pool_ego_metrics = attach_paradigms(pool_ego_metrics, metrics)
pool_mol_metrics = attach_paradigms(pool_mol_metrics, metrics)
pool_prot_metrics = attach_paradigms(pool_prot_metrics, metrics)
pool_lang_metrics = attach_paradigms(pool_lang_metrics, metrics)

print("✅ Each pool now contains:")
print("   • X features (12 descriptors)")
print("   • y_structural (original circular labels)")
print("   • y_functional (new exogenous labels)")

# ------------------------------------------------------------
# === DATA COLLECTION CONTAINERS ===
# ------------------------------------------------------------
rf_ranks_accum = []
gb_ranks_accum = []
lr_ranks_accum = []

all_run_results = []
all_run_cms = []
all_run_importance = []
all_run_similarity = []
pairwise_accumulator = {} 

# ------------------------------------------------------------
# === MAIN MULTI-SEED LOOP ===
# ------------------------------------------------------------
print(f"\n🚀 Starting benchmarking over {len(target_seeds)} seeds...")

for run_idx, s in enumerate(target_seeds):
    print(f"🌱 [SEED {s}] Processing {run_idx+1}/{len(target_seeds)}...")
    set_global_seed(s)

    # Pool sampling - we use 'label' to represent the DOMAIN CLASS 
    # (0: Social, 1: Molecular, 2: Proteins, 3: Linguistic).
    # Note: This is distinct from the anomaly labels used in Module 
    
    # --- DYNAMIC BALANCED SAMPLING ---
    # We identify the smallest pool to ensure perfectly balanced classes.
    # This prevents 'Class Imbalance Bias' in the domain classification task,
    # which is crucial for a rigorous Borda ranking.
    n_samples = min(500, 
                    len(pool_ego_metrics), 
                    len(pool_mol_metrics), 
                    len(pool_prot_metrics), 
                    len(pool_lang_metrics))
    
    if run_idx == 0:
        print(f"📊 Balanced sample size set to: {n_samples} graphs per domain.")

    # Sampling using the synchronized n_samples
    df_ego = pool_ego_metrics.sample(n_samples, random_state=s).copy()
    df_ego["label"] = 0

    df_qm9 = pool_mol_metrics.sample(n_samples, random_state=s).copy()
    df_qm9["label"] = 1

    df_prot = pool_prot_metrics.sample(n_samples, random_state=s).copy()
    df_prot["label"] = 2

    df_lang = pool_lang_metrics.sample(n_samples, random_state=s).copy()
    df_lang["label"] = 3

    # ------------------------------------------------------------
    # === INDUCTIVE TRAIN/TEST SPLIT FOR ANCHOR DISCOVERY ===
    # ------------------------------------------------------------
    df_ego_train, df_ego_test = train_test_split(
        df_ego, test_size=0.20, random_state=s
    )
    df_qm9_train, df_qm9_test = train_test_split(
        df_qm9, test_size=0.20, random_state=s
    )
    df_prot_train, df_prot_test = train_test_split(
        df_prot, test_size=0.20, random_state=s
    )
    df_lang_train, df_lang_test = train_test_split(
        df_lang, test_size=0.20, random_state=s
    )
    
    # Plot examples (first seed) - Modified to show NORMATIVE instances for clarity
    if run_idx == 0:

        idx_soc = pool_ego_metrics[
            pool_ego_metrics["y_functional"] == 0
        ].index[0]

        idx_mol = pool_mol_metrics[
            pool_mol_metrics["y_functional"] == 0
        ].index[0]

        idx_prot = pool_prot_metrics[
            pool_prot_metrics["y_functional"] == 0
        ].index[0]

        idx_lang = pool_lang_metrics[
            pool_lang_metrics["y_functional"] == 0
        ].index[0]

        graphs_info = [
            (
                "Social",
                pool_ego_graphs[idx_soc],
                DOMAIN_COLORS["Social"]
            ),
            (
                "Molecular",
                pool_mol_graphs[idx_mol],
                DOMAIN_COLORS["Molecular"]
            ),
            (
                "Proteins",
                pool_prot_graphs[idx_prot],
                DOMAIN_COLORS["Proteins"]
            ),
            (
                "Linguistic",
                pool_lang_graphs[idx_lang],
                DOMAIN_COLORS["Linguistic"]
            )
        ]

        fig_examples = plot_example_graphs_composite(
            graphs_info,
            filename="fig1_representative_domain_graphs"
        )

    plt.show()
    plt.close(fig_examples)

    df_all = pd.concat(
        [df_ego_train, df_qm9_train, df_prot_train, df_lang_train],
        ignore_index=True
    )
        
    X = df_all[metrics] # Keep raw data for splitting
    y = df_all["label"]

    # Multi-model Baseline
    models = {
        "RandomForest": RandomForestClassifier(n_estimators=300, max_depth=8, random_state=s),
        "GradientBoosting": GradientBoostingClassifier(random_state=s),
        "LogisticRegression": LogisticRegression(max_iter=1000, multi_class="multinomial", random_state=s)
    }
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=s)
    
    # Feature Importance (Borda) - Initialize container for the current seed
    feature_ranks_seed = pd.DataFrame() 

    for name, model in models.items():
        accs, f1s, aucs = [], [], []
        cms_model = np.zeros((4,4))
        # Store feature importance from each training fold for inductive
        # seed-level anchor ranking.
        fold_importances = [] 
        
        for tr, te in cv.split(X, y):
            # --- PREVENT DATA LEAKAGE ---
            # 1. Split raw data first
            X_train_fold, X_test_fold = X.iloc[tr], X.iloc[te]
            y_train_fold, y_test_fold = y.iloc[tr], y.iloc[te]

            # 2. Fit imputer and scaler ONLY on training fold (Strict Inductive Protocol)
            imputer_fold = SimpleImputer(strategy="median").fit(X_train_fold)
            X_train_imputed = imputer_fold.transform(X_train_fold)
            
            scaler_fold = StandardScaler().fit(X_train_imputed)
            
            # 3. Transform both folds using training statistics
            X_train_final = scaler_fold.transform(X_train_imputed)
            X_test_final = scaler_fold.transform(imputer_fold.transform(X_test_fold))

            # 4. Train model on processed training fold
            model.fit(X_train_final, y_train_fold)
            
            # 5. Predict and evaluate on processed test fold
            y_pred = model.predict(X_test_final)
            y_proba = model.predict_proba(X_test_final)
            
            accs.append(accuracy_score(y_test_fold, y_pred))
            f1s.append(f1_score(y_test_fold, y_pred, average="macro"))
            aucs.append(roc_auc_score(pd.get_dummies(y_test_fold), y_proba, multi_class="ovr"))
            
            if name == "GradientBoosting": 
                cms_model += (
                confusion_matrix(
                    y_test_fold,
                    y_pred,
                    labels=[0,1,2,3],
                    normalize="true"
                ) * 100
            )

            # --- INDUCTIVE FEATURE IMPORTANCE EXTRACTION ---
            # Feature importance is extracted only from the training fold.
            # The corresponding validation fold does not contribute to anchor ranking.
            if hasattr(model, "feature_importances_"):
                fi = model.feature_importances_
            else:
                # For linear models like Logistic Regression: average absolute coefficients
                fi = np.abs(model.coef_).mean(axis=0)
            fold_importances.append(fi)

        # Store global results for this seed
        all_run_results.append({"Model": name, "Accuracy": np.mean(accs), "F1_macro": np.mean(f1s), "ROC_AUC": np.mean(aucs)})
        if name == "GradientBoosting": all_run_cms.append(cms_model / 5)

        # --- STATISTICAL AGGREGATION FOR BORDA ---
        # Average fold-wise feature importance to obtain a seed-level estimate
        # for the Borda ranking.
        avg_fi_seed = np.mean(fold_importances, axis=0)
        feature_ranks_seed[name] = pd.Series(avg_fi_seed, index=metrics)

    # After evaluating all models for the current seed, compute the Borda Ranks
    # 1 = most important, 12 = least important
    seed_ranks = feature_ranks_seed.rank(ascending=False)
    
    # Update global accumulators for Table 4 (Consensus Ranking)
    rf_ranks_accum.append(seed_ranks["RandomForest"])
    gb_ranks_accum.append(seed_ranks["GradientBoosting"])
    lr_ranks_accum.append(seed_ranks["LogisticRegression"])
    
    # Store the seed-level consensus rank for downstream aggregation and visualization.
    all_run_importance.append(seed_ranks.mean(axis=1))

    # --- STRUCTURAL SIMILARITY ANALYSIS ---
    # We maintain the global similarity check based on domain centroids
    domains_dict = {
        "Social": df_ego_train,
        "Molecular": df_qm9_train,
        "Proteins": df_prot_train,
        "Linguistic": df_lang_train
    }
    domain_centroids = pd.DataFrame({d: domains_dict[d][metrics].mean() for d in domains_dict})
    all_run_similarity.append(domain_centroids.corr(method="pearson"))
    
    # Pairwise structural compatibility analysis using independent source
    # and target training samples.
    for (src_n, src_df), (tgt_n, tgt_df) in permutations(domains_dict.items(), 2):
        pair_key = f"{src_n} → {tgt_n}"
        if pair_key not in pairwise_accumulator: 
            pairwise_accumulator[pair_key] = {"corrs": [], "diffs": []}
        
        # Imputation and scaling are fitted on source + target training data only.
        combined_data = pd.concat(
            [src_df[metrics], tgt_df[metrics]],
            ignore_index=True
        )

        imputer_pair = SimpleImputer(strategy="median").fit(combined_data)

        combined_imputed = imputer_pair.transform(combined_data)

        scaler_temp = StandardScaler().fit(combined_imputed)

        src_imputed = imputer_pair.transform(src_df[metrics])
        tgt_imputed = imputer_pair.transform(tgt_df[metrics])

        src_scaled = pd.DataFrame(
            scaler_temp.transform(src_imputed),
            columns=metrics
        )

        tgt_scaled = pd.DataFrame(
            scaler_temp.transform(tgt_imputed),
            columns=metrics
        )
        
        sims = []
        for m in metrics:
            s_mu, t_mu = src_scaled[m].mean(), tgt_scaled[m].mean()
            s_std, t_std = src_scaled[m].std(), tgt_scaled[m].std()
            # Similarity based on standardized mean proximity relative to the
            # combined source-target dispersion.
            overlap_score = np.exp(-abs(s_mu - t_mu) / (np.sqrt(s_std**2 + t_std**2) + 1e-9))
            sims.append(overlap_score)
            
        diffs = [abs(src_scaled[m].mean() - tgt_scaled[m].mean()) for m in metrics]
        pairwise_accumulator[pair_key]["corrs"].append(sims)
        pairwise_accumulator[pair_key]["diffs"].append(diffs)

# ------------------------------------------------------------
# === FINAL AGGREGATION & RESULTS ===
# ------------------------------------------------------------
print("\n📊 Aggregating results and generating figures...")

# 1. Performance Comparison
# Each row in all_run_results corresponds to one model-level mean
# computed across the five cross-validation folds for one random seed.
# The final mean and standard deviation are therefore computed across
# the ten seed-level estimates.
df_perf_seed = pd.DataFrame(all_run_results)

df_perf = (
    df_perf_seed
    .groupby("Model")[["Accuracy", "F1_macro", "ROC_AUC"]]
    .agg(["mean", "std"])
)

print("\n=== Cross-Validated Performance Comparison (10-Seed Mean ± SD) ===")
display(df_perf)

df_perf_paper = pd.DataFrame({
    "Model": df_perf.index,
    "Accuracy": [
        f"{df_perf.loc[m, ('Accuracy', 'mean')]:.3f} ± "
        f"{df_perf.loc[m, ('Accuracy', 'std')]:.3f}"
        for m in df_perf.index
    ],
    "F1-macro": [
        f"{df_perf.loc[m, ('F1_macro', 'mean')]:.3f} ± "
        f"{df_perf.loc[m, ('F1_macro', 'std')]:.3f}"
        for m in df_perf.index
    ],
    "ROC-AUC": [
        f"{df_perf.loc[m, ('ROC_AUC', 'mean')]:.3f} ± "
        f"{df_perf.loc[m, ('ROC_AUC', 'std')]:.3f}"
        for m in df_perf.index
    ]
})

print("\n=== TABLE FOR MANUSCRIPT ===")
display(df_perf_paper)

df_perf_paper.to_csv(
    os.path.join(
        output_dir,
        "table_performance_10seed_mean_sd.csv"
    ),
    index=False
)
# ------------------------------------------------------------
# === CONFUSION MATRIX: MEAN ROW-NORMALIZED PERCENTAGES ===
# ------------------------------------------------------------

cm_avg = np.mean(all_run_cms, axis=0)

fig, ax = plt.subplots(
    figsize=(4.0, 3.6),
    constrained_layout=True
)

heatmap = sns.heatmap(
    cm_avg,
    annot=True,
    fmt=".2f",
    cmap="Blues",
    vmin=0,
    vmax=100,
    square=True,
    linewidths=0.5,
    linecolor="white",
    xticklabels=["Social", "Molecular", "Proteins", "Linguistic"],
    yticklabels=["Social", "Molecular", "Proteins", "Linguistic"],
    annot_kws={"fontsize": 7},
    ax=ax,
    cbar=False
)

ax.set_xlabel(
    "Predicted domain",
    fontsize=8,
    fontweight="normal"
)

ax.set_ylabel(
    "True domain",
    fontsize=8,
    fontweight="normal"
)

ax.tick_params(
    axis="both",
    labelsize=7,
    width=0.8
)

# Colorbar aligned with the heatmap
from mpl_toolkits.axes_grid1 import make_axes_locatable

divider = make_axes_locatable(ax)

cax = divider.append_axes(
    "right",
    size="4%",
    pad=0.08
)

cbar = fig.colorbar(
    heatmap.collections[0],
    cax=cax
)

cbar.set_label(
    "Percentage (%)",
    fontsize=7,
    labelpad=5
)

cbar.ax.tick_params(
    labelsize=6,
    width=0.7
)

sns.despine(
    ax=ax,
    top=True,
    right=True
)

save_highres(
    "fig_confusion_matrix",
    fig=fig
)

plt.show()
plt.close(fig)

# 3. Borda Feature Importance
avg_borda = pd.concat(all_run_importance, axis=1).mean(axis=1).sort_values()
df_importance = avg_borda.reset_index().rename(columns={"index": "feature", 0: "Borda_score"}).sort_values("Borda_score", ascending=False)
plt.figure(figsize=(8,4)); sns.barplot(data=df_importance, x="Borda_score", y="feature", color="skyblue")
plt.title("Borda Aggregated Feature Importance (Mean Rank)"); save_highres("borda_importance"); plt.show()

# ------------------------------------------------------------
# === TABLE CONSENSUS FEATURE IMPORTANCE (BORDA COUNT) ===
# ------------------------------------------------------------
print("\n📊 Table Consensus Feature Importance Ranking (Borda Count)")

df_table_4 = pd.DataFrame({
    "Feature": metrics,
    "RF Rank": pd.concat(rf_ranks_accum, axis=1).mean(axis=1).values,
    "GB Rank": pd.concat(gb_ranks_accum, axis=1).mean(axis=1).values,
    "LR Rank": pd.concat(lr_ranks_accum, axis=1).mean(axis=1).values
})

df_table_4["Borda Score"] = df_table_4[["RF Rank", "GB Rank", "LR Rank"]].mean(axis=1)
df_table_4 = df_table_4.sort_values("Borda Score").reset_index(drop=True)
display(df_table_4.style.format({"RF Rank": "{:.1f}", "GB Rank": "{:.1f}", "LR Rank": "{:.1f}", "Borda Score": "{:.3f}"}))
df_table_4.to_csv(os.path.join(output_dir, "table_4_consensus_ranking.csv"), index=False)

# 4. Structural Similarity Matrix
domain_order = ["Social", "Molecular", "Proteins", "Linguistic"]
avg_sim = pd.concat(all_run_similarity).groupby(level=0).mean()
avg_sim = avg_sim.reindex(index=domain_order, columns=domain_order)
print("\n🧠 Structural Similarity Matrix (Averaged):")
display(avg_sim.round(3))
plt.figure(figsize=(6,5)); sns.heatmap(avg_sim, annot=True, cmap="coolwarm", vmin=0, vmax=1)
plt.title("Structural Similarity Matrix — Domain-Level"); save_highres("structural_similarity_matrix"); plt.show()

# ------------------------------------------------------------------------------
# === PAIRWISE ANALYSIS & GLOBAL CONSENSUS IIT_score (G) CALCULATION ===
# ------------------------------------------------------------------------------
final_pairwise = []
all_iit_vectors = [] 

# Aligned Borda reference
df_importance = pd.concat(all_run_importance, axis=1).mean(axis=1).reset_index()
df_importance.columns = ["feature", "Borda_score"]
df_importance = df_importance.set_index("feature").reindex(metrics).reset_index()

for p_key, data in pairwise_accumulator.items():
    mean_corrs = np.mean(data["corrs"], axis=0)
    mean_diffs = np.mean(data["diffs"], axis=0)
    
    # Pairwise IIT score calculation
    # Structural compatibility is symmetric with respect to source and target;
    # transfer directionality is evaluated in downstream experiments.
    iit_vector_pairwise = df_importance["Borda_score"].values * (np.power(mean_corrs, 2) / (1 + mean_diffs))
    all_iit_vectors.append(iit_vector_pairwise)
    
    df_tmp = pd.DataFrame({"feature": metrics, "score": iit_vector_pairwise})
    top_8_pair = df_tmp.sort_values("score", ascending=False).head(8)
    
    final_pairwise.append({
        "DomainPair": p_key, 
        "IIT_score_pair": round(top_8_pair["score"].mean(), 4), # Mean IIT score of the selected top-eight features for the pair.
        "top_transferable_feats": top_8_pair["feature"].tolist()
    })

# Aggregate each feature pairwise IIT score across all source-target pairs
# to obtain the global consensus score.
df_importance["IIT_score_G"] = np.mean(all_iit_vectors, axis=0)

# Select the eight general candidate structural anchors with the highest
# global consensus IIT scores.
transferable_features = df_importance.sort_values("IIT_score_G", ascending=False).head(8)["feature"].tolist()

# Update global variables for Module 3 & 4
df_pw = pd.DataFrame(final_pairwise)
globals()["df_pairwise_transfer"] = df_pw
globals()["transferable_features"] = transferable_features

# ------------------------------------------------------------
# === EXPORT COMPLETE PAIRWISE TRANSFER TABLE ===
# ------------------------------------------------------------

# Display complete feature lists without pandas truncation
pd.set_option("display.max_colwidth", None)

print("\n📊 Pairwise structural bridge strength (IIT scores):")
display(df_pw)

# Save complete pairwise transfer table
pairwise_csv_path = os.path.join(
    output_dir,
    "table_pairwise_transfer_complete.csv"
)

df_pw.to_csv(
    pairwise_csv_path,
    index=False
)

print(f"💾 Complete pairwise transfer table saved to: {pairwise_csv_path}")

# ------------------------------------------------------------------------------
# === VISUALIZATION: FIGURE 2 (Global Ranked Structural Anchors) ===
# ------------------------------------------------------------------------------
df_ranked = df_importance.sort_values("IIT_score_G", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(
    figsize=(7.0, 5.6),
    constrained_layout=True
)
ax = plt.gca()

# Highlight the eight selected candidate anchors separately from the remaining features.
colors_bar = [
    "#0072B2" if i < 8 else "#BDBDBD"
    for i in range(len(df_ranked))
]

sns.barplot(x="IIT_score_G", y="feature", data=df_ranked, palette=colors_bar, edgecolor='#333333', linewidth=0.8)

from matplotlib.ticker import FormatStrFormatter
ax.xaxis.set_major_formatter(FormatStrFormatter('%.2f')) 
ax.tick_params(
    axis="x",
    labelsize=7,
    width=0.8
)
ax.tick_params(
    axis="y",
    labelsize=7,
    width=0.8
)   

# Visual separator between the top-eight and remaining features.
cutoff_position = 7.5

ax.axhline(
    y=cutoff_position,
    color="#333333",
    linestyle="--",
    linewidth=1.0,
    alpha=0.8
)

ax.text(
    df_ranked["IIT_score_G"].max() * 0.98,
    cutoff_position,
    "Top-8 anchor threshold",
    color="#333333",
    fontsize=7,
    fontweight="normal",
    ha="right",
    va="bottom"
)

ax.set_xlabel(
    r"Global consensus $\mathrm{IIT}_{score,G}$",
    fontsize=8,
    fontweight="normal"
)
ax.set_ylabel("")

sns.despine(
    ax=ax,
    top=True,
    right=True,
    left=False,
    bottom=False
)

save_highres(
    "fig2_global_ranked_anchors",
    fig=fig
)
plt.show() 

# ------------------------------------------------------------
# === FINALIZE GLOBALS ===
# ------------------------------------------------------------

globals()["df_importance"] = df_importance
globals()["transferable_features"] = transferable_features
globals()["df_pairwise_transfer"] = df_pw
globals()["G_social_full"], globals()["dataset_qm9"], globals()["dataset_prot"], globals()["lang_pool"] = G_fb_full, dataset_qm9, dataset_prot, pool_lang_graphs

globals()["df_ego_train"] = df_ego_train
globals()["df_qm9_train"] = df_qm9_train
globals()["df_prot_train"] = df_prot_train
globals()["df_lang_train"] = df_lang_train

globals()["df_ego_test"] = df_ego_test
globals()["df_qm9_test"] = df_qm9_test
globals()["df_prot_test"] = df_prot_test
globals()["df_lang_test"] = df_lang_test

print(f"\n✅ MODULE 1 Completed.")
print(f"📌 Global Structural Anchors identified: {transferable_features}")
print(f"📌 Analysis averaged over {len(target_seeds)} independent seeds.")
```

```python
# ==============================================================================
# === MODULE 1 DIAGNOSTICS: Functional vs. Structural Paradigm Validation ===
# ==============================================================================
# Objective: Characterize the new functional labels (y_functional) and compare
# them with the original structural labels (y_structural) to document how the
# two validation paradigms differ. The functional labels are defined from
# domain-specific properties or controlled perturbations rather than from the
# topological descriptors used for structural anchor discovery.
#
# These analyses are diagnostic only: they do not establish statistical
# independence or prove the absence of all possible topological associations.
# Their purpose is to document label distributions, task-specific properties,
# and the separation between the structural and functional validation paradigms.
# ------------------------------------------------------------------------------

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import mannwhitneyu, wilcoxon

print("🔍 MODULE 1 DIAGNOSTICS: RIGOROUS PARADIGM VALIDATION\n" + "="*60)

# 1. LABEL DISTRIBUTION AND OVERLAP DIAGNOSTIC
# Descriptive comparison of the structural-verification and functional/
# perturbation-based labels. Label overlap does not by itself establish
# independence from the topological descriptor space.
diag_rows = []
domains_dict = {
    "Social": pool_ego_metrics,
    "Molecular": pool_mol_metrics,
    "Proteins": pool_prot_metrics,
    "Linguistic": pool_lang_metrics
}

for name, df in domains_dict.items():
    count_struct = df["y_structural"].sum()
    count_func = df["y_functional"].sum()
    
    # Calculate Jaccard-like overlap: (Common Anomalies) / (Total Unique Anomalies)
    common = ((df["y_structural"] == 1) & (df["y_functional"] == 1)).sum()
    total_unique = (df["y_structural"] | df["y_functional"]).sum()
    overlap_pct = (common / total_unique * 100) if total_unique > 0 else 0
    
    diag_rows.append({
        "Domain": name, 
        "Total Observations": len(df),
        "Structural Anom (N)": count_struct,
        "Functional Anom (N)": count_func,
        "Common Count": common,
        "Overlap Percentage": f"{overlap_pct:.2f}%"
    })

print("📊 TABLE: Structural vs. Functional/Perturbation Label Overlap")
display(pd.DataFrame(diag_rows))

# 2. TOPOLOGICAL SENSITIVITY ANALYSIS (Efficiency & Assortativity)
# Testing if the primary structural anchors can sense exogenous functional shifts.
domains_list = [
    ("Social", pool_ego_metrics),
    ("Molecular", pool_mol_metrics),
    ("Proteins", pool_prot_metrics),
    ("Linguistic", pool_lang_metrics)
]

for metric_name in ["efficiency", "assortativity"]:
    print(f"\n🔍 BOXPLOT ANALYSIS: {metric_name.upper()} Sensitivity")
    fig, axes = plt.subplots(
        2, 4,
        figsize=(7.0, 4.0),
        constrained_layout=True)
    
    for i, (name, df) in enumerate(domains_list):
        sns.boxplot(
            data=df,
            x="y_structural",
            y=metric_name,
            palette="Set2",
            ax=axes[0, i]) 
        axes[0, i].set_title(
            f"{name}: Structural",
            fontsize=8)

        sns.boxplot(
            data=df,
            x="y_functional",
            y=metric_name,
            palette="Set1",
            ax=axes[1, i]) 
        axes[1, i].set_title(
            f"{name}: Functional",
            fontsize=8)
        
        # Section 2: Boxplot Analysis Update
        # Removed the "Pending" text because Social anomalies are now generated in Module 1
        pass

    plt.tight_layout()
    plt.show()

# 3. NON-PARAMETRIC STATISTICAL VALIDATION (Mann-Whitney U)
# Quantifying the separability of Normal vs. Anomalous instances for both paradigms.
print("\n🧪 STATISTICAL SIGNIFICANCE: Topological Anchors as Functional Proxies")
topo_stats = []
for m in ["efficiency", "assortativity"]:
    for name, df in domains_list:
        for p_name, p_col in [("Structural", "y_structural"), ("Functional", "y_functional")]:
            g0 = df[df[p_col] == 0][m].dropna()
            g1 = df[df[p_col] == 1][m].dropna()
            if len(g0) > 0 and len(g1) > 0:
                stat, p = mannwhitneyu(g0, g1, alternative='two-sided')
                topo_stats.append({
                    "Domain": name, "Paradigm": p_name, "Anchor": m, 
                    "U-Stat": stat, "p-value": p
                })

display(pd.DataFrame(topo_stats).style.format({"p-value": "{:.2e}"}))

# 4. LABEL-CONSTRUCTION AND CONTROL DIAGNOSTICS
# Sanity checks for the variables used to define functional or controlled-
# perturbation labels. These analyses do not constitute independent validation
# of the labels.
print("\n🔍 PHYSICAL VALIDATION: Distribution of Exogenous Properties")
fig, axes = plt.subplots(
    2, 2,
    figsize=(7.0, 7.0),
    constrained_layout=True)
axes = axes.flatten()

# --- 1. MOLECULAR: Chemical Energy Gap ---
ax = axes[0]
stable_vals = [v for v in mol_external_gap if v >= gap_threshold]
unstable_vals = [v for v in mol_external_gap if v < gap_threshold]
sns.kdeplot(data=stable_vals, fill=True, color="#95a5a6", label="Stable (Norm)", ax=ax)
sns.kdeplot(data=unstable_vals, fill=True, color="#e74c3c", label="Unstable (Anom)", ax=ax)
ax.set_title("Molecular: HOMO-LUMO Gap (eV)", fontweight="normal")
ax.legend(frameon=False)

# --- 2. PROTEINS: SIZE-CONFOUND CHECK ---
ax = axes[1]
prot_norm_size = pool_prot_metrics[pool_prot_metrics["y_functional"] == 0]["n_nodes"]
prot_anom_size = pool_prot_metrics[pool_prot_metrics["y_functional"] == 1]["n_nodes"]
sns.kdeplot(data=prot_norm_size, fill=True, label="Enzymes", color="#27AE60", ax=ax)
sns.kdeplot(data=prot_anom_size, fill=True, label="Non-Enzymes", color="#c0392b", ax=ax)
ax.set_title("Proteins: Physical Size (Amino Acids)", fontweight="normal")
ax.legend(frameon=False)

# --- 3. LINGUISTIC: Sentence Length Control ---
ax = axes[2]
lang_norm_nodes = pool_lang_metrics[pool_lang_metrics["y_functional"] == 0]["n_nodes"]
lang_anom_nodes = pool_lang_metrics[pool_lang_metrics["y_functional"] == 1]["n_nodes"]
sns.kdeplot(data=lang_norm_nodes, fill=True, label="Natural", color="#4682B4", ax=ax)
sns.kdeplot(data=lang_anom_nodes, fill=True, label="Scrambled", color="#F39C12", ax=ax)
ax.set_title("Linguistic: Sentence Length (Control)", fontweight="normal")
ax.legend(frameon=False)

# --- 4. SOCIAL: PERTURBATION SANITY CHECK ---
# Confirm that the controlled fragmentation procedure changes connectivity
# while preserving the number of nodes.
ax = axes[3]
soc_norm_edges = pool_ego_metrics[pool_ego_metrics["y_functional"] == 0]["n_edges"]
soc_anom_edges = pool_ego_metrics[pool_ego_metrics["y_functional"] == 1]["n_edges"]
sns.kdeplot(data=soc_norm_edges, fill=True, label="Cohesive (Norm)", color="#34495e", ax=ax)
sns.kdeplot(data=soc_anom_edges, fill=True, label="Fragmented (Anom)", color="#9b59b6", ax=ax)
ax.set_title("Social: Edge Density (Fragmentation Signal)", fontweight="normal")
ax.legend(frameon=False)

plt.tight_layout()
plt.show()

# 5. FINAL EXOGENOUS VALIDATION SUMMARY (REVISED)
soc_norm_nodes = pool_ego_metrics[pool_ego_metrics["y_functional"] == 0]["n_nodes"]
soc_anom_nodes = pool_ego_metrics[pool_ego_metrics["y_functional"] == 1]["n_nodes"]

exo_final_results = [
    {
        "Domain": "Molecular", "Property": "HOMO-LUMO Gap", 
        "p-value": mannwhitneyu(stable_vals, unstable_vals)[1], "Target": "Label-definition check"
    },
    {
        "Domain": "Proteins", "Property": "Physical Size", 
        "p-value": mannwhitneyu(prot_norm_size, prot_anom_size)[1], "Target": "p < 0.05"
    },
    {
        "Domain": "Linguistic", "Property": "Sentence Length (Control)", 
        "p-value": mannwhitneyu(lang_norm_nodes, lang_anom_nodes)[1], "Target": "p > 0.05"
    },
    {
        "Domain": "Social", "Property": "User Count (Control)", 
        "p-value": mannwhitneyu(soc_norm_nodes, soc_anom_nodes)[1], "Target": "p > 0.05"
    }
]

df_exo_final = pd.DataFrame(exo_final_results)
display(df_exo_final.style.format({"p-value": "{:.2e}"}))

print("\n📝 DIAGNOSTIC SUMMARY:")

for idx, row in df_exo_final.iterrows():
    p = row["p-value"]
    prop = row["Property"]
    domain = row["Domain"]
    
    if "Control" in prop:
        # For Social and Linguistic, we want to prove the groups are physically identical
        if p > 0.05:
            print(f"Domain {domain}: ✅ No evidence of a {prop.lower()} difference (p={p:.2f})")
            print("   -> Interpretation: The diagnostic does not detect a significant size difference.")
        else:
            print(f"Domain {domain}: ⚠️ Evidence of a {prop.lower()} difference (p={p:.2e})")
            print("   -> Interpretation: A potential size-related confound requires consideration.")
    else:
        # For Molecular and Proteins, we want to prove the groups are physically distinct
        if p < 0.05:
            print(f"Domain {domain}: ✅ SUCCESS (Significant Physical Signal) (p={p:.2e})")
            print(f"   -> Interpretation: Low p-value confirms that normal and anomalous groups are physically distinct.")
        else:
            print(f"Domain {domain}: ❌ FAILED (Insufficient Signal) (p={p:.2f})")

# ==============================================================================
# === MODULE 1 DIAGNOSTICS: Linguistic Perturbation Sanity Check ===
# ==============================================================================
# Objective: Assess whether word-order scrambling changes a corpus-based
# linguistic likelihood measure. This serves as a perturbation sanity check
# rather than an independent exogenous validation because the reference model
# derives from the same Brown News corpus used for graph construction.
# ------------------------------------------------------------------------------

from collections import Counter
import math

print("\n🧪 EXOGENOUS LINGUISTIC VALIDATION: Syntax Integrity Check")

# 1. Build a Reference Bigram Model from the News Corpus
# We learn what "Normal English" looks like from the entire News category
news_words = [w.lower() for w in brown.words(categories='news') if w.isalpha()]
bigrams = zip(news_words, news_words[1:])
bigram_counts = Counter(bigrams)
total_bigrams = sum(bigram_counts.values())

def get_sentence_probability(word_list):
    """Calculates the average bigram probability of a sentence."""
    if len(word_list) < 2: return 0
    probs = []
    for i in range(len(word_list)-1):
        pair = (word_list[i], word_list[i+1])
        # Probability of the pair occurring (with Laplace smoothing)
        prob = (bigram_counts[pair] + 1) / (total_bigrams + len(set(news_words)))
        probs.append(math.log(prob))
    return sum(probs) / len(probs) # Average Log-Likelihood

# 2. Calculate Integrity Score for Natural vs Scrambled sentences
# We re-extract the raw words from our pool_lang_graphs (which we saved in Mod 1)
# or we use a fresh sample for this check
print("   • Scoring sentence integrity (Log-Likelihood)...")

# We sample directly from the News sentences we used
natural_sents = [s for s in brown.sents(categories='news') if 10 <= len([w for w in s if w.isalpha()]) <= 30][:200]
scores_natural = [get_sentence_probability([w.lower() for w in s if w.isalpha()]) for s in natural_sents]

# Create scrambled versions
scores_scrambled = []
rng_diag = random.Random(MASTER_POOL_SEED)
for s in natural_sents:
    words = [w.lower() for w in s if w.isalpha()]
    rng_diag.shuffle(words)
    scores_scrambled.append(get_sentence_probability(words))

# 3. Statistical Comparison
stat_ling, p_ling = wilcoxon(
    scores_natural,
    scores_scrambled,
    alternative="two-sided"
)

print(f"\nAnalysis of Linguistic Integrity (Syntactic Likelihood):")
print(f"  • Mean Score (Natural):   {np.mean(scores_natural):.2f} (Higher = More Natural)")
print(f"  • Mean Score (Scrambled): {np.mean(scores_scrambled):.2f} (Lower = More Random)")
print(f"  • Statistical Significance (p-value): {p_ling:.2e}")

# 4. Final Verification Plot
plt.figure(figsize=(10, 6))
sns.kdeplot(scores_natural, fill=True, label="Natural (Normative)", color="#4682B4")
sns.kdeplot(scores_scrambled, fill=True, label="Scrambled (Anomalous)", color="#F39C12")
plt.title("Linguistic Validation: Syntactic Integrity Distribution", fontweight="normal")
plt.xlabel("Log-Likelihood Score (Exogenous NLP Metric)")
plt.ylabel("Density")
plt.legend()
plt.show()

if p_ling < 0.05:
    print("✅ SUCCESS: Scrambled sentences differ from their natural counterparts")
    print("   on the corpus-based linguistic likelihood measure.")
else:
    print("❌ FAILED: The scrambling did not destroy the linguistic signal enough.")

# ==============================================================================
# === NEW: SOCIAL FRAGMENTATION VALIDATION (Exogenous Check) ===
# ==============================================================================
print("\n🔍 SOCIAL VALIDATION: Verifying Connectivity Collapse (Exogenous Signal)")

# Extract edge counts for the two social groups
soc_norm_edges = pool_ego_metrics[pool_ego_metrics["y_functional"] == 0]["n_edges"]
soc_anom_edges = pool_ego_metrics[pool_ego_metrics["y_functional"] == 1]["n_edges"]

# Statistical check for the difference in connectivity
stat_soc, p_soc = mannwhitneyu(soc_norm_edges, soc_anom_edges)

plt.figure(figsize=(8, 5))
sns.kdeplot(data=soc_norm_edges, fill=True, label="Cohesive (Natural)", color="#34495e")
sns.kdeplot(data=soc_anom_edges, fill=True, label="Fragmented (Anomalous)", color="#9b59b6")
plt.title("Social Domain: Evidence of Functional Fragmentation", fontweight="normal")
plt.xlabel("Number of Edges (Connectivity)")
plt.ylabel("Density")
plt.legend(frameon=False)
plt.show()

print(f"Social Connectivity Shift (p-value): {p_soc:.2e}")
if p_soc < 0.05:
    print("✅ SUCCESS: The fragmented group differs in connectivity as expected.")
    print("   This confirms that the controlled edge-removal perturbation was applied.")
```

# MODULE 3 — Manifold Synchronization & Targeted Stress Trajectory

```python
# ==============================================================================
# === X-CDTL Framework | Module 3: Cross-Domain Transfer & Stress Robustness ===
# ==============================================================================
# Researcher: Daniele Caligiore
# Affiliation: CNR - Italian National Research Council (ISTC)
# Version: 2.0 (Sep 2026)
# ------------------------------------------------------------------------------
# DESCRIPTION:
# This module evaluates cross-domain anomaly-detection transfer using the
# candidate structural anchors identified in Module 1.
#
# The analysis follows an inductive evaluation protocol in which anchor
# selection, preprocessing, stress generation, and dimensionality reduction
# use training observations only. A clean held-out target test set is kept
# unchanged until final evaluation.
#
# Four stress regimes independently vary target-data availability and feature
# noise. Performance is compared across target-only, sample-count-matched
# target-only, and source-to-target transfer conditions using ROC-AUC,
# average precision, and F1 score.
#
# The functional-validation paradigm uses domain-specific functional labels
# or controlled perturbation labels defined in Module 1. The structural-
# verification paradigm retains the original structural labels for comparison.
#
# CORE PIPELINE:
# 1. Identification of pair-specific candidate structural anchors from Module 1.
# 2. Reproducible sampling and inductive train/test splitting for source and
#    target domains.
# 3. Target-training stress generation through data scarcity, feature noise,
#    and controlled missingness.
# 4. Source-plus-target joint preprocessing and PCA for transfer, compared
#    with target-only preprocessing and PCA.
# 5. Optional IIT-derived feature weighting followed by Isolation Forest
#    anomaly detection.
# 6. Evaluation on the clean held-out target test set and aggregation of
#    performance across domain pairs, seeds, and stress regimes.
# ------------------------------------------------------------------------------

# 1. SYSTEM IMPORTS FIRST
import os, subprocess

# ------------------------------------------------------------
# === 1. RIGOROUS GPU SELECTION & ENVIRONMENT SETUP ===
# ------------------------------------------------------------
# CRITICAL: We set CUDA visibility at the OS level BEFORE importing torch
# to ensure the backend respects the environment variable.
def get_optimal_gpu():
    try:
        smi_output = subprocess.check_output(
            "nvidia-smi --query-gpu=index,name,memory.free,memory.total --format=csv,noheader,nounits",
            shell=True
        ).decode("utf-8").strip().split("\n")
        gpus = [line.split(",") for line in smi_output]
        # Sort by free memory (index 2) descending
        gpus = [(int(i), n.strip(), int(f)) for i, n, f in gpus]
        gpus.sort(key=lambda x: x[2], reverse=True)
        return str(gpus[0][0])
    except:
        return "0"

os.environ["CUDA_VISIBLE_DEVICES"] = get_optimal_gpu()
print(f"🌍 OS level: Targeted GPU set to index {os.environ['CUDA_VISIBLE_DEVICES']}")

# 2. HEAVY LIBRARIES IMPORTS (Now it's safe)
import torch
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.ensemble import IsolationForest
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    f1_score,
    precision_score,
    recall_score
)

# ------------------------------------------------------------
# === 2. PYTORCH & CUDA INITIALIZATION ===
# ------------------------------------------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

if device.type == "cuda":
    gpu_name = torch.cuda.get_device_name(0)
    free_mem, total_mem = torch.cuda.mem_get_info()
    print(f"🖥️ PyTorch initialized on GPU: {gpu_name}")
    print(f"📊 VRAM Status: {free_mem//1024**2} MB free / {total_mem//1024**2} MB total")
else:
    print("⚠️ GPU not available, running on CPU")

# ------------------------------------------------------------
# === 3. ALGORITHM SELECTION (RAPIDS cuML vs Scikit-Learn) ===
# ------------------------------------------------------------
try:
    from cuml.ensemble import IsolationForest as cuIsolationForest
    from cuml.neighbors import LocalOutlierFactor as cuLOF
    use_cuml = True
    print("🚀 cuML Native CUDA acceleration enabled.")
except ImportError:
    use_cuml = False
    from sklearn.neighbors import LocalOutlierFactor
    print("⚠️ cuML not available — falling back to standard scikit-learn.")

# Selection of the Algorithm Class
USE_LOF = False

# ------------------------------------------------------------
# === ANOMALY PREVALENCE CONFIGURATION (PRIORS) ===
# ------------------------------------------------------------
# Set the Isolation Forest contamination prior according to the active
# validation paradigm. Structural verification uses the fixed 0.10 prior
# associated with the original structural-label definition, whereas
# functional validation uses domain-specific label-prevalence priors.
STRUCTURAL_PRIOR = 0.10

# For Functional Validation, we use fixed priors based on dataset characteristics.
# Proteins have a ~45% non-enzyme prevalence in the 10-30 node range.
FUNCTIONAL_PRIORS = {
    "Molecular": 0.10,
    "Social": 0.10,
    "Linguistic": 0.10,
    "Proteins": 0.45 
}


STRESS_TRAJECTORY = [
    (0.9, 0.1),  # 1. Baseline
    (0.1, 0.1),  # 2. Data scarcity
    (0.9, 0.9),  # 3. Feature noise
    (0.1, 0.9)   # 4. Combined stress
]

# Ensure the paradigm switch is inherited from Module 1
if 'VALIDATION_PARADIGM' not in globals():
    VALIDATION_PARADIGM = "functional_validation"

print(f"🚀 Module 3 initialized. Active Paradigm: {VALIDATION_PARADIGM.upper()}")

# ------------------------------------------------------------
# === AUTOMATIC FEATURE WEIGHTING CONFIGURATION ===
# ------------------------------------------------------------
# For functional validation, pair-specific candidate anchors are further
# weighted using the IIT-derived source-target similarity term computed
# from training observations only. Structural verification retains uniform
# weighting for consistency with the structural anomaly definition.

USE_WEIGHTED_TRANSFER = True

print(f"⚙️ Feature Weighting active: {USE_WEIGHTED_TRANSFER}")

seeds_to_run = target_seeds 
missing_frac = 0.05
n_trees = 1024         # High-resolution manifold estimation
max_features = 1.0     # Full anchor evaluation
n_pca_comp = 8         # Maximum information retention

pool_map = {
    "df_qm9": "pool_mol_metrics",
    "df_ego": "pool_ego_metrics",
    "df_lang": "pool_lang_metrics",
    "df_prot": "pool_prot_metrics"
}

domain_pairs = [
    # Source: Social
    ("Social → Molecular", "df_ego", "df_qm9"),
    ("Social → Proteins", "df_ego", "df_prot"),
    ("Social → Linguistic", "df_ego", "df_lang"),
    # Source: Molecular
    ("Molecular → Social", "df_qm9", "df_ego"),
    ("Molecular → Proteins", "df_qm9", "df_prot"),
    ("Molecular → Linguistic", "df_qm9", "df_lang"),
    # Source: Proteins
    ("Proteins → Social", "df_prot", "df_ego"),
    ("Proteins → Molecular", "df_prot", "df_qm9"),
    ("Proteins → Linguistic", "df_prot", "df_lang"),
    # Source: Linguistic
    ("Linguistic → Social", "df_lang", "df_ego"),
    ("Linguistic → Molecular", "df_lang", "df_qm9"),
    ("Linguistic → Proteins", "df_lang", "df_prot")
]

# ------------------------------------------------------------
# === CONTAMINATION-SENSITIVITY ANALYSIS ====================
# ------------------------------------------------------------
# The main functional-validation operating point is domain-specific:
# 0.10 for Molecular, Social, and Linguistic domains, and 0.45 for
# Proteins. The sensitivity analysis evaluates a common grid of
# contamination values across all domain pairs.
# ------------------------------------------------------------

CONTAMINATION_LEVELS = np.array(
    [0.05, 0.10, 0.20, 0.30, 0.45]
)

# ------------------------------------------------------------
# === UTILITIES (CORRUPT & METRICS) ===
# ------------------------------------------------------------
def corrupt_dataframe(df, noise_std=0.0, missing_frac=0.0, rng=None):
    dfc = df.copy().astype(float)

    if noise_std > 0:
        dfc += rng.normal(
            0,
            noise_std * (dfc.std() + 1e-6),
            size=dfc.shape
        )

    if missing_frac > 0:
        mask = rng.rand(*dfc.shape) < missing_frac
        dfc[mask] = np.nan

    # Imputation is performed later by the scenario-specific
    # training-only preprocessing pipeline.
    return dfc

def safe_metrics(y_true, scores, y_pred):
    out = {
        "roc_auc": np.nan,
        "ap": np.nan,
        "f1": np.nan,
        "precision": np.nan,
        "recall": np.nan
    }

    if len(np.unique(y_true)) > 1:
        try:
            out["roc_auc"] = roc_auc_score(
                y_true,
                scores
            )
        except:
            pass

        try:
            out["ap"] = average_precision_score(
                y_true,
                scores
            )
        except:
            pass

    try:
        out["f1"] = f1_score(
            y_true,
            y_pred,
            zero_division=0
        )
    except:
        pass

    try:
        out["precision"] = precision_score(
            y_true,
            y_pred,
            zero_division=0
        )
    except:
        pass

    try:
        out["recall"] = recall_score(
            y_true,
            y_pred,
            zero_division=0
        )
    except:
        pass

    return out

# ------------------------------------------------------------
# === CORE EXPERIMENT LOOP (POOL-AWARE) ===
# ------------------------------------------------------------
def run_transfer_aligned_trajectory_optimized(
    src_pool_name,
    tgt_pool_name,
    domain_label
):
    """
    Executes the cross-domain transfer learning pipeline using an inductive
    protocol. Anchor discovery relies on training observations, while the
    clean target test set remains untouched until final evaluation.
    """

    # ------------------------------------------------------------
    # 1. IDENTIFY PAIR-SPECIFIC STRUCTURAL ANCHORS
    # ------------------------------------------------------------
    feats_row = df_pairwise_transfer[
        df_pairwise_transfer["DomainPair"] == domain_label
    ]

    feats = (
        feats_row["top_transferable_feats"].iloc[0]
        if not feats_row.empty
        else transferable_features
    )

    IForestClass = (
        cuIsolationForest
        if (use_cuml and device.type == "cuda")
        else IsolationForest
    )

    results = []
    contamination_results = []

    # ------------------------------------------------------------
    # 2. MULTI-SEED EXPERIMENT
    # ------------------------------------------------------------
    for s in seeds_to_run:
        set_global_seed(s)

        # --------------------------------------------------------
        # INDUCTIVE STEP 1:
        # Reproduce the same 500-graph sampling and 80/20 split
        # used in Module 1.
        # --------------------------------------------------------
        tgt_pool_full = globals()[pool_map[tgt_pool_name]]
        src_pool_full = globals()[pool_map[src_pool_name]]

        target_sample = tgt_pool_full.sample(
            500,
            random_state=s
        ).copy()

        target_train_pool, target_test_pool = train_test_split(
            target_sample,
            test_size=0.20,
            random_state=s
        )

        source_sample = src_pool_full.sample(
            500,
            random_state=s
        ).copy()

        source_train_pool, _ = train_test_split(
            source_sample,
            test_size=0.20,
            random_state=s
        )

        # --------------------------------------------------------
        # Clean held-out target test set.
        # It is not used for sampling, stress generation, preprocessing fitting,
        # anchor weighting, PCA fitting, or anomaly-detector training.
        # The test observations are transformed only after all training-dependent
        # steps have been fitted.
        # --------------------------------------------------------
        X_test_raw = target_test_pool[feats].copy()

        if VALIDATION_PARADIGM == "functional_validation":
            y_test = target_test_pool["y_functional"].values
        else:
            # Structural labels were already computed in Module 1.
            # Reusing them avoids redefining labels from the test features.
            y_test = target_test_pool["y_structural"].values

        # --------------------------------------------------------
        # 3. TARGETED STRESS TRAJECTORY
        # --------------------------------------------------------
        for frac, nl in STRESS_TRAJECTORY:

            # ----------------------------------------------------
            # INDUCTIVE STEP 2:
            # Subsample ONLY the target training pool.
            # The 100 held-out target test observations remain fixed.
            # ----------------------------------------------------
            n_target_train = max(
                1,
                int(round(len(target_train_pool) * frac))
            )

            target_train_subset = target_train_pool.sample(
                n=n_target_train,
                random_state=s
            ).copy()

            X_train_raw = target_train_subset[feats].copy()

            # ----------------------------------------------------
            # Apply noise and missingness to TRAINING data only.
            # ----------------------------------------------------
            rng = np.random.RandomState(s)

            X_train_stressed = corrupt_dataframe(
                X_train_raw,
                noise_std=nl,
                missing_frac=missing_frac,
                rng=rng
            )

            # Source remains clean and comes only from source training data.
            Xs_raw = source_train_pool[feats].copy()

            # ====================================================
            # TRANSFER PIPELINE
            # ====================================================

            # ----------------------------------------------------
            # INDUCTIVE STEP 3A:
            # Imputation and scaling fitted on Source + Target
            # training observations only.
            # ----------------------------------------------------
            combined_train_raw = pd.concat(
                [Xs_raw, X_train_stressed],
                ignore_index=True
            )

            imputer_transfer = SimpleImputer(
                strategy="median"
            ).fit(combined_train_raw)

            scaler_transfer = StandardScaler().fit(
                imputer_transfer.transform(combined_train_raw)
            )

            Xs_scaled = pd.DataFrame(
                scaler_transfer.transform(
                    imputer_transfer.transform(Xs_raw)
                ),
                columns=feats
            )

            X_train_scaled = pd.DataFrame(
                scaler_transfer.transform(
                    imputer_transfer.transform(X_train_stressed)
                ),
                columns=feats
            )

            X_test_transfer_scaled = pd.DataFrame(
                scaler_transfer.transform(
                    imputer_transfer.transform(X_test_raw)
                ),
                columns=feats
            )

            # ----------------------------------------------------
            # INDUCTIVE STEP 4A:
            # IIT-based feature weighting using training data only.
            # ----------------------------------------------------
            if USE_WEIGHTED_TRANSFER:

                weights = []

                for m in feats:

                    mu_s = Xs_scaled[m].mean()
                    mu_t = X_train_scaled[m].mean()

                    std_s = Xs_scaled[m].std()
                    std_t = X_train_scaled[m].std()

                    w = np.exp(
                        -abs(mu_s - mu_t) /
                        (
                            np.sqrt(
                                std_s**2 + std_t**2
                            ) + 1e-9
                        )
                    )

                    weights.append(w**2)

                w_vector = (
                    np.array(weights) /
                    np.mean(weights)
                )

                Xs_scaled *= w_vector
                X_train_scaled *= w_vector
                X_test_transfer_scaled *= w_vector

            # ----------------------------------------------------
            # INDUCTIVE STEP 5A:
            # PCA fitted only on source + target training data.
            # ----------------------------------------------------
            pca_transfer = PCA(
                n_components=n_pca_comp,
                random_state=s
            )

            pca_transfer.fit(
                pd.concat(
                    [Xs_scaled, X_train_scaled],
                    ignore_index=True
                )
            )

            Xs_aligned = pca_transfer.transform(
                Xs_scaled
            )

            X_train_transfer_aligned = pca_transfer.transform(
                X_train_scaled
            )

            X_test_transfer_aligned = pca_transfer.transform(
                X_test_transfer_scaled
            )

            # ====================================================
            # NO-TRANSFER TARGET-ONLY PIPELINE
            # ====================================================

            # ----------------------------------------------------
            # INDUCTIVE STEP 3B:
            # Imputation and scaling fitted on target training data only.
            # ----------------------------------------------------
            imputer_nt = SimpleImputer(
                strategy="median"
            ).fit(X_train_stressed)

            scaler_nt = StandardScaler().fit(
                imputer_nt.transform(X_train_stressed)
            )

            X_train_nt_scaled = pd.DataFrame(
                scaler_nt.transform(
                    imputer_nt.transform(X_train_stressed)
                ),
                columns=feats
            )

            X_test_nt_scaled = pd.DataFrame(
                scaler_nt.transform(
                    imputer_nt.transform(X_test_raw)
                ),
                columns=feats
            )

            # ----------------------------------------------------
            # INDUCTIVE STEP 4B:
            # Target-only PCA.
            # ----------------------------------------------------
            pca_nt = PCA(
                n_components=n_pca_comp,
                random_state=s
            )

            pca_nt.fit(X_train_nt_scaled)

            X_train_nt_aligned = pca_nt.transform(
                X_train_nt_scaled
            )

            X_test_nt_aligned = pca_nt.transform(
                X_test_nt_scaled
            )

            # ====================================================
            # EXPERIMENTAL SCENARIOS
            # ====================================================

            n_target = len(X_train_nt_aligned)
            n_source = len(Xs_aligned)

            # Sample-count matching:
            # replicate target-only training observations so that the control
            # contains the same number of training observations as the transfer
            # condition, without introducing additional observations.
            n_matched = n_source + n_target

            n_repeats = int(
                np.ceil(n_matched / n_target)
            )

            X_train_augmented = np.tile(
                X_train_nt_aligned,
                (n_repeats, 1)
            )[:n_matched]

            # Safety check: the matched control must contain exactly the same
            # number of training observations as the transfer condition.
            assert len(X_train_augmented) == n_matched

            scenarios = {
                "No Transfer (Aligned)": (
                    X_train_nt_aligned,
                    X_test_nt_aligned
                ),

                "Transfer (Aligned)": (
                    np.vstack([
                        Xs_aligned,
                        X_train_transfer_aligned
                    ]),
                    X_test_transfer_aligned
                ),

                "No Transfer (Augmented)": (
                    X_train_augmented,
                    X_test_nt_aligned
                )
            }

            # ----------------------------------------------------
            # 6. CONTAMINATION PRIOR
            # ----------------------------------------------------
            if VALIDATION_PARADIGM == "structural_verification":

                current_prior = STRUCTURAL_PRIOR

            else:

                target_key = (
                    "Molecular"
                    if "mol" in tgt_pool_name
                    else "Proteins"
                    if "prot" in tgt_pool_name
                    else "Social"
                    if "ego" in tgt_pool_name
                    else "Linguistic"
                )

                current_prior = FUNCTIONAL_PRIORS[target_key]

            # ----------------------------------------------------
            # 7. FIT AND EVALUATE ALL SCENARIOS
            # ----------------------------------------------------
            for sc, (X_tr, X_te) in scenarios.items():

                clf = IForestClass(
                    n_estimators=n_trees,
                    contamination=current_prior,
                    max_features=max_features,
                    random_state=s,
                    bootstrap=True
                )

                clf.fit(X_tr)

                # Evaluation uses the clean held-out target test set.
                scores = -clf.decision_function(X_te)

                y_pred = (
                    clf.predict(X_te) == -1
                ).astype(int)

                m = safe_metrics(
                    y_test,
                    scores,
                    y_pred
                )

                m.update({
                    "DomainPair": domain_label,
                    "Scenario": sc,
                    "Seed": s,
                    "low_data_frac": frac,
                    "noise_level": nl
                })

                results.append(m)

                # ------------------------------------------------
                # CONTAMINATION-SENSITIVITY ANALYSIS
                # ------------------------------------------------
                # Re-fit the same detector on the same training
                # data for each contamination level. The clean
                # held-out target test set remains unchanged.
                #
                # The sample-count-matched augmented control is
                # excluded because it addresses training-set-size
                # effects rather than contamination sensitivity.
                # ------------------------------------------------

                if sc in [
                    "No Transfer (Aligned)",
                    "Transfer (Aligned)"
                ]:

                    for contamination_level in CONTAMINATION_LEVELS:

                        clf_sens = IForestClass(
                            n_estimators=n_trees,
                            contamination=float(
                                contamination_level
                            ),
                            max_features=max_features,
                            random_state=s,
                            bootstrap=True
                        )

                        clf_sens.fit(X_tr)

                        scores_sens = (
                            -clf_sens.decision_function(X_te)
                        )

                        y_pred_sens = (
                            clf_sens.predict(X_te) == -1
                        ).astype(int)

                        m_sens = safe_metrics(
                            y_test,
                            scores_sens,
                            y_pred_sens
                        )

                        m_sens.update({
                            "DomainPair": domain_label,
                            "Scenario": sc,
                            "Seed": s,
                            "low_data_frac": frac,
                            "noise_level": nl,
                            "Contamination": float(
                                contamination_level
                            )
                        })

                        contamination_results.append(m_sens)

    

                

    # --------------------------------------------------------
    # RETURN AFTER COMPLETING ALL SEEDS, STRESS LEVELS,
    # AND SCENARIOS FOR THIS DOMAIN PAIR
    # --------------------------------------------------------
    contamination_results_df = (
        pd.DataFrame(contamination_results)
        if contamination_results
        else pd.DataFrame()
    )

    return (
        pd.DataFrame(results),
        contamination_results_df
    )

# ------------------------------------------------------------
# === EXECUTION: TARGETED STRESS TRAJECTORY ===
# ------------------------------------------------------------
print(f"\n🚀 Running Targeted Stress Trajectory ({len(seeds_to_run)} seeds x 4 stress levels)...")

all_trajectory_results = []
all_contamination_results = []

for label, src, tgt in domain_pairs:

    print(f"   • Processing trajectory: {label}")

    res, contamination_res = (
        run_transfer_aligned_trajectory_optimized(
            src,
            tgt,
            label
        )
    )

    all_trajectory_results.append(res)

    if not contamination_res.empty:
        all_contamination_results.append(
            contamination_res
        )

df_aligned = pd.concat(
    all_trajectory_results,
    ignore_index=True
)

df_contamination = pd.concat(
    all_contamination_results,
    ignore_index=True
)


# ------------------------------------------------------------
# 1. DEFINE THE FOUR STRESS REGIMES
# ------------------------------------------------------------

def assign_stress_regime(row):

    frac = row["low_data_frac"]
    noise = row["noise_level"]

    if frac == 0.9 and noise == 0.1:
        return "Baseline"

    if frac == 0.1 and noise == 0.1:
        return "Data Scarcity"

    if frac == 0.9 and noise == 0.9:
        return "Feature Noise"

    if frac == 0.1 and noise == 0.9:
        return "Combined Stress"

    return "Other"


df_contamination["Regime"] = (
    df_contamination.apply(
        assign_stress_regime,
        axis=1
    )
)

# ------------------------------------------------------------
# 2. BUILD CLEAN TRANSFER VS NO-TRANSFER TABLE
# ------------------------------------------------------------
# Keep only the two directly comparable scenarios.
#
# Each row remains identified by:
#   DomainPair × Seed × Regime × Contamination
#
# Therefore no averaging occurs at this stage.

contamination_pairwise_raw = (
    df_contamination[
        df_contamination["Scenario"].isin([
            "No Transfer (Aligned)",
            "Transfer (Aligned)"
        ])
    ]
    .pivot_table(
        index=[
            "DomainPair",
            "Seed",
            "Regime",
            "low_data_frac",
            "noise_level",
            "Contamination"
        ],
        columns="Scenario",
        values=[
            "f1",
            "precision",
            "recall"
        ],
        aggfunc="mean"
    )
)

# ------------------------------------------------------------
# 3. FLATTEN MULTI-INDEX COLUMNS
# ------------------------------------------------------------

contamination_pairwise_raw.columns = [
    f"{metric}_{scenario}"
    .replace(" ", "_")
    .replace("(", "")
    .replace(")", "")
    for metric, scenario in contamination_pairwise_raw.columns
]

contamination_pairwise_raw = (
    contamination_pairwise_raw
    .reset_index()
)

# ------------------------------------------------------------
# 4. COMPUTE TRANSFER-ASSOCIATED CHANGES
# ------------------------------------------------------------

contamination_pairwise_raw["F1_Gain"] = (
    contamination_pairwise_raw[
        "f1_Transfer_Aligned"
    ]
    -
    contamination_pairwise_raw[
        "f1_No_Transfer_Aligned"
    ]
)

contamination_pairwise_raw["Precision_Gain"] = (
    contamination_pairwise_raw[
        "precision_Transfer_Aligned"
    ]
    -
    contamination_pairwise_raw[
        "precision_No_Transfer_Aligned"
    ]
)

contamination_pairwise_raw["Recall_Gain"] = (
    contamination_pairwise_raw[
        "recall_Transfer_Aligned"
    ]
    -
    contamination_pairwise_raw[
        "recall_No_Transfer_Aligned"
    ]
)

# ------------------------------------------------------------
# 5. OUTPUT 1:
#    COMBINED STRESS — PAIRWISE
#
#    12 domain pairs × 6 contamination levels 
#    Mean ± SD across the 10 experimental seeds.
# ------------------------------------------------------------

print("Available columns:", contamination_pairwise_raw.columns.tolist())

combined_pairwise = (
    contamination_pairwise_raw[
        contamination_pairwise_raw["Regime"]
        == "Combined Stress"
    ]
    .groupby(
        [
            "DomainPair",
            "Contamination"
        ]
    )[
        [
            "f1_No_Transfer_Aligned",
            "f1_Transfer_Aligned",
            "F1_Gain",
            "precision_No_Transfer_Aligned",
            "precision_Transfer_Aligned",
            "Precision_Gain",
            "recall_No_Transfer_Aligned",
            "recall_Transfer_Aligned",
            "Recall_Gain"
        ]
    ]
    .agg(
        ["mean", "std"]
    )
    .reset_index()
)

# Flatten columns for a publication-friendly CSV.
combined_pairwise.columns = [
    "_".join(
        [str(x) for x in col if str(x) != ""]
    )
    if isinstance(col, tuple)
    else str(col)
    for col in combined_pairwise.columns
]

print(
    "\n📌 OUTPUT 1 — COMBINED STRESS, PAIRWISE"
)

display(
    combined_pairwise.round(3)
)

combined_pairwise_output = os.path.join(
    output_dir,
    f"contamination_combined_pairwise_{VALIDATION_PARADIGM}.csv"
)

combined_pairwise.to_csv(
    combined_pairwise_output,
    index=False
)

print(
    f"💾 Pairwise combined-stress contamination results exported to:\n"
    f"   {combined_pairwise_output}"
)

# ------------------------------------------------------------
# 6. OUTPUT 2:
#    COMBINED STRESS — GLOBAL
#
#    All 12 domain pairs × 10 seeds.
#    Mean ± SD across all pair-seed observations.
#
#    This provides the global contamination-sensitivity summary,
#    while Output 1 preserves the pairwise structure.
# ------------------------------------------------------------

combined_global = (
    contamination_pairwise_raw[
        contamination_pairwise_raw["Regime"]
        == "Combined Stress"
    ]
    .groupby(
        ["Contamination"]
    )[
        [
            "f1_No_Transfer_Aligned",
            "f1_Transfer_Aligned",
            "F1_Gain",
            "precision_No_Transfer_Aligned",
            "precision_Transfer_Aligned",
            "Precision_Gain",
            "recall_No_Transfer_Aligned",
            "recall_Transfer_Aligned",
            "Recall_Gain"
        ]
    ]
    .agg(
        ["mean", "std"]
    )
    .reset_index()
)

combined_global.columns = [
    "_".join(
        [str(x) for x in col if str(x) != ""]
    )
    if isinstance(col, tuple)
    else str(col)
    for col in combined_global.columns
]

print(
    "\n📌 OUTPUT 2 — COMBINED STRESS, GLOBAL"
)

display(
    combined_global.round(3)
)

combined_global_output = os.path.join(
    output_dir,
    f"contamination_combined_global_{VALIDATION_PARADIGM}.csv"
)

combined_global.to_csv(
    combined_global_output,
    index=False
)

print(
    f"💾 Global combined-stress contamination results exported to:\n"
    f"   {combined_global_output}"
)

# ------------------------------------------------------------
# 7. OUTPUT 3:
#    FOUR STRESS REGIMES — PAIRWISE
#
#    For each:
#       DomainPair × Regime × Contamination
#
#    mean ± SD across the 10 experimental seeds.
#
#    This preserves the pairwise structure while allowing direct
#    comparison of baseline, data scarcity, feature noise, and
#    combined stress.
# ------------------------------------------------------------

all_regimes_pairwise = (
    contamination_pairwise_raw
    .groupby(
        [
            "DomainPair",
            "Regime",
            "Contamination"
        ]
    )[
        [
            "f1_No_Transfer_Aligned",
            "f1_Transfer_Aligned",
            "F1_Gain",
            "precision_No_Transfer_Aligned",
            "precision_Transfer_Aligned",
            "Precision_Gain",
            "recall_No_Transfer_Aligned",
            "recall_Transfer_Aligned",
            "Recall_Gain"
        ]
    ]
    .agg(
        ["mean", "std"]
    )
    .reset_index()
)

all_regimes_pairwise.columns = [
    "_".join(
        [str(x) for x in col if str(x) != ""]
    )
    if isinstance(col, tuple)
    else str(col)
    for col in all_regimes_pairwise.columns
]

print(
    "\n📌 OUTPUT 3 — FOUR STRESS REGIMES, PAIRWISE"
)

display(
    all_regimes_pairwise.round(3)
)

all_regimes_output = os.path.join(
    output_dir,
    f"contamination_all_regimes_pairwise_{VALIDATION_PARADIGM}.csv"
)

all_regimes_pairwise.to_csv(
    all_regimes_output,
    index=False
)

print(
    f"💾 Four-regime pairwise contamination results exported to:\n"
    f"   {all_regimes_output}"
)

# ------------------------------------------------------------
# 8. ADDITIONAL GLOBAL FOUR-REGIME SUMMARY
#
#    This is intentionally secondary to Output 3.
#    It averages over all 12 domain pairs and 10 seeds.
# ------------------------------------------------------------

all_regimes_global = (
    contamination_pairwise_raw
    .groupby(
        [
            "Regime",
            "Contamination"
        ]
    )[
        [
            "f1_No_Transfer_Aligned",
            "f1_Transfer_Aligned",
            "F1_Gain",
            "precision_No_Transfer_Aligned",
            "precision_Transfer_Aligned",
            "Precision_Gain",
            "recall_No_Transfer_Aligned",
            "recall_Transfer_Aligned",
            "Recall_Gain"
        ]
    ]
    .agg(
        ["mean", "std"]
    )
    .reset_index()
)

all_regimes_global.columns = [
    "_".join(
        [str(x) for x in col if str(x) != ""]
    )
    if isinstance(col, tuple)
    else str(col)
    for col in all_regimes_global.columns
]

print(
    "\n📌 ADDITIONAL GLOBAL FOUR-REGIME SUMMARY"
)

display(
    all_regimes_global.round(3)
)

all_regimes_global_output = os.path.join(
    output_dir,
    f"contamination_all_regimes_global_{VALIDATION_PARADIGM}.csv"
)

all_regimes_global.to_csv(
    all_regimes_global_output,
    index=False
)

print(
    f"💾 Global four-regime contamination results exported to:\n"
    f"   {all_regimes_global_output}"
)

# ------------------------------------------------------------
# 9. OUTPUT 4:
#    COMBINED-STRESS FIGURE
#
#    Transfer-associated F1 change:
#        Transfer F1 - No Transfer F1
#
#    Each line represents one directed domain pair.
#    Values are mean ± SD across the 10 experimental seeds.
# ------------------------------------------------------------

combined_plot = (
    contamination_pairwise_raw[
        contamination_pairwise_raw["Regime"]
        == "Combined Stress"
    ]
    .groupby(
        [
            "DomainPair",
            "Contamination"
        ]
    )["F1_Gain"]
    .agg(
        ["mean", "std"]
    )
    .reset_index()
)

plt.figure(
    figsize=(8.0, 6.0)
)

for domain_pair in domain_pairs:

    label = domain_pair[0]

    sub = (
        combined_plot[
            combined_plot["DomainPair"] == label
        ]
        .sort_values("Contamination")
    )

    if sub.empty:
        continue

    x = sub["Contamination"].to_numpy()
    y = sub["mean"].to_numpy()
    sd = sub["std"].to_numpy()

    plt.plot(
        x,
        y,
        marker="o",
        linewidth=1.2,
        markersize=4,
        label=label
    )

    plt.fill_between(
        x,
        y - sd,
        y + sd,
        alpha=0.08
    )

plt.axhline(
    0.0,
    linewidth=0.8,
    linestyle="--"
)

plt.xlabel(
    "Isolation Forest contamination"
)

plt.ylabel(
    "Transfer-associated change in F1 score"
)

plt.xticks(
    CONTAMINATION_LEVELS,
    [
        f"{int(x * 100)}%"
        for x in CONTAMINATION_LEVELS
    ]
)

plt.legend(
    frameon=False,
    fontsize=6,
    ncol=2
)

sns.despine()

save_highres(
    f"contamination_sensitivity_combined_pairwise_{VALIDATION_PARADIGM}"
)

plt.show()
plt.close()

print(
    "\n✅ Detailed contamination-sensitivity reporting completed."
)

# ------------------------------------------------------------
# === CONTAMINATION-SENSITIVITY SUMMARY ===
# ------------------------------------------------------------

print(
    "\n📊 CONTAMINATION-SENSITIVITY ANALYSIS "
    "(clean held-out target test sets):"
)

contamination_summary = (
    df_contamination
    .groupby(
        [
            "low_data_frac",
            "noise_level",
            "Scenario",
            "Contamination"
        ]
    )[
        ["f1", "precision", "recall"]
    ]
    .agg(["mean", "std"])
)

display(
    contamination_summary.round(3)
)

contamination_output = os.path.join(
    output_dir,
    f"contamination_sensitivity_{VALIDATION_PARADIGM}.csv"
)

df_contamination.to_csv(
    contamination_output,
    index=False
)

print(
    "\n🔎 CHECK: main functional-validation operating points"
)

main_operating_points = {
    "Molecular": 0.10,
    "Social": 0.10,
    "Linguistic": 0.10,
    "Proteins": 0.45
}

for target_domain, operating_point in main_operating_points.items():

    check_main = df_contamination[
        df_contamination["Contamination"].eq(
            operating_point
        )
        &
        df_contamination["DomainPair"].str.contains(
            target_domain,
            regex=False
        )
    ]

    print(
        f"\n{target_domain}: contamination = {operating_point:.2f}"
    )

    display(
        check_main.groupby(
            ["DomainPair", "Scenario"]
        )["f1"]
        .agg(["mean", "std"])
        .round(3)
    )

print(
    f"💾 Contamination-sensitivity results exported to: "
    f"{contamination_output}"
)

# ------------------------------------------------------------
# === CONTAMINATION-SENSITIVITY PLOT ===
# ------------------------------------------------------------

combined_contamination = (
    df_contamination[
        df_contamination["low_data_frac"].eq(0.1)
        & df_contamination["noise_level"].eq(0.9)
        & df_contamination["Scenario"].isin([
            "No Transfer (Aligned)",
            "Transfer (Aligned)"
        ])
    ]
    .groupby(
        ["Contamination", "Scenario"]
    )["f1"]
    .agg(["mean", "std"])
    .reset_index()
)

plt.figure(figsize=(5.0, 3.5))

for scenario in [
    "No Transfer (Aligned)",
    "Transfer (Aligned)"
]:

    sub = combined_contamination[
        combined_contamination["Scenario"] == scenario
    ].sort_values("Contamination")

    x = sub["Contamination"].to_numpy()
    y = sub["mean"].to_numpy()
    sd = sub["std"].to_numpy()

    plt.plot(
        x,
        y,
        marker="o",
        linewidth=1.5,
        label=scenario
    )

    plt.fill_between(
        x,
        y - sd,
        y + sd,
        alpha=0.15
    )

plt.xlabel("Isolation Forest contamination")
plt.ylabel("F1 score")
plt.xticks(
    CONTAMINATION_LEVELS,
    [
        f"{int(x * 100)}%"
        for x in CONTAMINATION_LEVELS
    ]
)

plt.legend(
    frameon=False,
    fontsize=7
)

sns.despine()

save_highres(
    f"contamination_sensitivity_combined_{VALIDATION_PARADIGM}"
)

plt.show()
plt.close()

# ------------------------------------------------------------
# === FINAL AGGREGATION: STRATIFIED PERFORMANCE REPORT ===
# ------------------------------------------------------------
print(f"\n📊 Generating Stratified Performance Report for: {VALIDATION_PARADIGM.upper()}")

# 1. Map the 4 logical scenarios to the raw alpha/eta parameters
def map_regime_names(row):
    a, n = row['low_data_frac'], row['noise_level']
    if a == 0.9 and n == 0.1: return "1. Baseline"
    if a == 0.1 and n == 0.1: return "2. Data Scarcity"
    if a == 0.9 and n == 0.9: return "3. Feature Noise"
    if a == 0.1 and n == 0.9: return "4. Combined Stress"
    return "Other"

df_aligned['Regime'] = df_aligned.apply(map_regime_names, axis=1)

# 2. GLOBAL SUMMARY: Performance per Regime and Scenario
# This table shows the 4 cases separately as you requested
global_stratified_stats = df_aligned.groupby(["Regime", "Scenario"])[["roc_auc", "ap", "f1"]].mean().round(3)

print("\n🌍 GLOBAL PERFORMANCE PER OPERATIONAL REGIME (Averaged over seeds & pairs):")
display(global_stratified_stats)

# 3. DETAILED STATISTICAL REPORT (Mean ± Std)
# This provides the uncertainty (SD) for each of the 4 cases
detailed_stats = df_aligned.groupby(["Regime", "Scenario"])[["roc_auc", "ap", "f1"]].agg(["mean", "std"])

# Format for publication (μ ± σ)
final_report_rows = []
for (regime, scenario), row in detailed_stats.iterrows():
    final_report_rows.append({
        "Regime": regime,
        "Scenario": scenario,
        "ROC_AUC (μ ± σ)": f"{row[('roc_auc', 'mean')]:.3f} ± {row[('roc_auc', 'std')]:.3f}",
        "AP (μ ± σ)": f"{row[('ap', 'mean')]:.3f} ± {row[('ap', 'std')]:.3f}",
        "F1 (μ ± σ)": f"{row[('f1', 'mean')]:.3f} ± {row[('f1', 'std')]:.3f}"
    })

df_final_table = pd.DataFrame(final_report_rows)
print("\n📈 # STRATIFIED PERFORMANCE SUMMARY (MEAN ± SD):")
display(df_final_table)

# 4. EXPORT RESULTS
output_filename = f"stratified_results_{VALIDATION_PARADIGM}.csv"
df_final_table.to_csv(os.path.join(output_dir, output_filename), index=False)
print(f"\n💾 Results exported to: {output_filename}")

if torch.cuda.is_available():
    f_mem, t_mem = torch.cuda.mem_get_info()
    print(f"\n✅ Module 3 completed. VRAM remaining: {f_mem//1024**2} MB")

print(
    f"\n✅ MODULE 3 completed across "
    f"{len(seeds_to_run)} random seeds."
)
```

```python
# ==============================================================================
# === X-CDTL Framework | Module 4: Predictive Validation & Transfer Gain ===
# ==============================================================================
# Researcher: Daniele Caligiore
# Affiliation: CNR - Italian National Research Council (ISTC)
# Version: 2.0 (Sep 2026)
# ------------------------------------------------------------------------------
# DESCRIPTION:
# This module evaluates the relationship between pairwise IIT compatibility
# and empirical transfer gain across the stress regimes evaluated in Module 3.
#
# Transfer gain is computed as the difference in performance between the
# source-to-target transfer condition and the target-only condition, using
# the metric selected for the active validation paradigm.
#
# The analysis is performed at the domain-pair level by averaging results
# across seeds. The resulting pair-level association is assessed using
# Spearman rank correlation with a two-sided permutation test.
# The same procedure is used for the exploratory intra-target analysis,
# where only three source domains are available for each target.
#
# Visualization emphasizes both seed-level variability and pair-level
# mean performance. The analyses are descriptive of the observed
# cross-domain associations and do not establish a general predictive law.
#
# CORE PIPELINE:
# 1. Extract seed-level transfer gains from Module 3.
# 2. Merge transfer gains with pairwise IIT scores from Module 1.
# 3. Aggregate the main correlation analysis at the domain-pair level.
# 4. Visualize the association across four stress regimes.
# 5. Perform exploratory intra-target analysis under combined stress using the same permutation-based Spearman test.
# ==============================================================================

from scipy.stats import spearmanr, linregress
import itertools
from statsmodels.stats.multitest import multipletests

print("\n📊 Running MODULE 4 — Predictive Validation & Transfer Gain Analysis")

# --- Output directory ---
if 'results_dir' not in globals():
    results_dir = "results_nature"

os.makedirs(results_dir, exist_ok=True)

plt.rcParams.update({
    "font.family": "sans-serif",
    "font.size": 8,
    "axes.labelsize": 8,
    "axes.titlesize": 8,
    "xtick.labelsize": 7,
    "ytick.labelsize": 7,
    "legend.fontsize": 7,
    "axes.linewidth": 0.8,
    "xtick.major.width": 0.8,
    "ytick.major.width": 0.8,
    "pdf.fonttype": 42,
    "ps.fonttype": 42
})

# --- 2️⃣ PERFORMANCE SUMMARY & DATA INTEGRITY ---
# We automatically adapt the analysis based on the global paradigm switch
if 'VALIDATION_PARADIGM' not in globals():
    VALIDATION_PARADIGM = "functional_validation"

# Target metric selection for TGI correlation:
# - Functional: F1-score (captures the decision stabilization/rescue effect)
# - Structural: ROC-AUC (captures mathematical outlier detection)
TGI_METRIC = "f1" if VALIDATION_PARADIGM == "functional_validation" else "roc_auc"

if 'df_aligned' not in globals():
    raise RuntimeError("❌ df_aligned not found. Please execute Module 3 first.")

print(f"📊 Module 4 Analysis: {VALIDATION_PARADIGM.upper()}")
print(f"🎯 Target metric for Transfer Gain Index (TGI): {TGI_METRIC.upper()}")
print(f"📈 Summarizing results across {len(target_seeds)} seeds...")

# --- 4️⃣ SEED-LEVEL DATA INTEGRATION ---
def extract_high_power_gains(df, metric="f1"):
    grouped = (
        df.groupby(
            ["DomainPair", "Seed",
             "low_data_frac", "noise_level", "Scenario"]
        )[metric]
        .mean()
        .unstack()
    )

    transfer_gain = (
        grouped["Transfer (Aligned)"]
        - grouped["No Transfer (Aligned)"]
    )

    return (
        transfer_gain
        .reset_index()
        .rename(columns={0: "TransferGain"})
    )

print("📈 Integrating Empirical Gains with Module 1 IIT Scores...")
df_gains_hp = extract_high_power_gains(df_aligned, TGI_METRIC)

df_final = df_gains_hp.merge(
    df_pairwise_transfer[["DomainPair", "IIT_score_pair"]],
    on="DomainPair",
    how="inner"
)

# --- 5️⃣ VISUALIZATION 1: PREDICTIVE VALIDATION ACROSS ABLATION SCENARIOS ---
def plot_robust_quadrants(df, filename):
    ablation_scenarios = [
        {"a": 0.9, "n": 0.1, "title": "1. Baseline"},
        {"a": 0.1, "n": 0.1, "title": "2. Data Scarcity Only"},
        {"a": 0.9, "n": 0.9, "title": "3. Feature Noise Only"},
        {"a": 0.1, "n": 0.9, "title": "4. Combined Stress"}
    ]
    fig, axes = plt.subplots(
        2, 2,
        figsize=(7.0, 5.8),
        constrained_layout=False
    )
    axes = axes.flatten()
    SOURCE_COLORS = {
        "Social": "#0072B2",
        "Molecular": "#CC79A7",
        "Proteins": "#E69F00",
        "Linguistic": "#56B4E9"}
    
    print("\n🧪 PAIR-LEVEL ASSOCIATION BETWEEN IIT AND TRANSFER GAIN:")

    association_results = []

    x_min = df["IIT_score_pair"].min()
    x_max = df["IIT_score_pair"].max()

    y_min = df["TransferGain"].min()
    y_max = df["TransferGain"].max()

    x_pad = 0.05 * (x_max - x_min)
    y_pad = 0.05 * (y_max - y_min)

    x_limits = (x_min - x_pad, x_max + x_pad)
    y_limits = (y_min - y_pad, y_max + y_pad)

    for i, config in enumerate(ablation_scenarios):
        ax = axes[i]
        ax.text(
            -0.18,
            1.08,
            chr(97 + i),
            transform=ax.transAxes,
            fontsize=9,
            fontweight="bold",
            va="top",
            ha="right"
        )
        
        sub = df[(df["low_data_frac"] == config["a"]) & (df["noise_level"] == config["n"])]
        # Statistical analysis at the domain-pair level:
        # seed-level transfer gains are averaged within each directed domain pair,
        # yielding one observation per pair for the main correlation analysis.

        sub_pair = (
            sub.groupby("DomainPair", as_index=False)
            .agg({
                "IIT_score_pair": "first",
                "TransferGain": "mean"
            })
        )

        r_s, p_perm, n_perm = spearman_permutation_test(
            sub_pair["IIT_score_pair"],
            sub_pair["TransferGain"]
        )

        association_results.append({
            "Scenario": config["title"],
            "low_data_frac": config["a"],
            "noise_level": config["n"],
            "n_pairs": len(sub_pair),
            "Spearman_r": r_s,
            "Permutation_p": p_perm,
            "N_permutations": n_perm
        })

        slope, intercept, _, _, _ = linregress(
            sub_pair["IIT_score_pair"],
            sub_pair["TransferGain"]
        )

        x_range = np.linspace(
            sub_pair["IIT_score_pair"].min() - 0.5,
            sub_pair["IIT_score_pair"].max() + 0.5,
            100
        )

        ax.plot(x_range, intercept + slope * x_range, color='black', lw=1.2, linestyle='--', alpha=0.5)

        for j, dp in enumerate(sub["DomainPair"].unique()):
            pair_data = sub[sub["DomainPair"] == dp]; src = dp.split(" → ")[0]

            ax.scatter(
                pair_data["IIT_score_pair"],
                pair_data["TransferGain"],
                color=SOURCE_COLORS.get(src),
                s=12,
                alpha=0.20,
                edgecolors="none",
                zorder=1
            )

            ax.scatter(
                pair_data["IIT_score_pair"].mean(),
                pair_data["TransferGain"].mean(),
                color=SOURCE_COLORS.get(src),
                marker="o",
                s=55,
                edgecolors="black",
                linewidth=0.8,
                zorder=3
            )

        ax.set_title(
            config["title"].replace("1. ", "")
            .replace("2. ", "")
            .replace("3. ", "")
            .replace("4. ", ""),
            fontsize=8,
            fontweight="normal",
            pad=4
        )
        ax.text(
            0.98,
            0.98,
            rf"$r_s$ = {r_s:.2f}, "
            rf"$p_{{perm}}$ = {p_perm:.3f}, "
            rf"$n$ = {len(sub_pair)}",
            transform=ax.transAxes,
            fontsize=7,
            va="top",
            ha="right"
        )

        ax.set_xlabel(r"Pairwise IIT score", fontsize=8, fontweight="normal")
        if i in [0, 2]:
            ax.set_ylabel(r"Transfer gain ($\Delta$F1)", fontsize=8, fontweight="normal")
        else:
            ax.set_ylabel("")

        ax.set_xlim(x_limits)
        ax.set_ylim(y_limits)
        
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)
        ax.tick_params(axis="both", which="major", labelsize=8, width=0.8)
        print(
            f"{config['title']}: n_pairs = {len(sub_pair)}, "
            f"r_s = {r_s:.3f} | "
            f"permutation p = {p_perm:.4f} | "
            f"permutations = {n_perm}"
        )

    # Save pair-level IIT–transfer association statistics
    df_association = pd.DataFrame(association_results)

    association_output = os.path.join(
        results_dir,
        f"iit_transfer_association_{VALIDATION_PARADIGM}.csv"
    )

    df_association.to_csv(
        association_output,
        index=False
    )

    print(
        f"\n📄 IIT–transfer association results saved to: "
        f"{association_output}"
    )

    from matplotlib.lines import Line2D

    source_handles = [
        Line2D(
            [0], [0],
            marker="o",
            linestyle="None",
            markerfacecolor=SOURCE_COLORS["Social"],
            markeredgecolor="black",
            markersize=5,
            label="Social"
        ),
        Line2D(
            [0], [0],
            marker="o",
            linestyle="None",
            markerfacecolor=SOURCE_COLORS["Molecular"],
            markeredgecolor="black",
            markersize=5,
            label="Molecular"
        ),
        Line2D(
            [0], [0],
            marker="o",
            linestyle="None",
            markerfacecolor=SOURCE_COLORS["Proteins"],
            markeredgecolor="black",
            markersize=5,
            label="Proteins"
        ),
        Line2D(
            [0], [0],
            marker="o",
            linestyle="None",
            markerfacecolor=SOURCE_COLORS["Linguistic"],
            markeredgecolor="black",
            markersize=5,
            label="Linguistic"
        )
    ]

    plt.subplots_adjust(
        left=0.13,
        right=0.98,
        bottom=0.29,
        top=0.95,
        wspace=0.25,
        hspace=0.50
    )

    fig.legend(
        handles=source_handles,
        loc="upper center",
        bbox_to_anchor=(0.5, -0.02),
        bbox_transform=fig.transFigure,
        ncol=4,
        frameon=False,
        fontsize=7
    )

    save_highres(filename); plt.show()

# ------------------------------------------------------------
# === VISUALIZATION 2: EXPLORATORY INTRA-TARGET ANALYSIS ===
# ------------------------------------------------------------
def plot_intra_target_validation(df, filename):
    df_rescue = df[(df["low_data_frac"] == 0.1) & (df["noise_level"] == 0.9)].copy()
    df_rescue['Target'] = df_rescue['DomainPair'].apply(lambda x: x.split(" → ")[1])
    
    fig, axes = plt.subplots(
        2, 2,
        figsize=(7.0, 5.8),
        constrained_layout=True)
    axes = axes.flatten()
    SOURCE_COLORS = {
        "Social": "#0072B2",
        "Molecular": "#CC79A7",
        "Proteins": "#E69F00",
        "Linguistic": "#56B4E9"}
    
    print("\n🧪 EXPLORATORY INTRA-TARGET ANALYSIS (Combined Stress):")
    for i, tgt in enumerate(["Molecular", "Proteins", "Social", "Linguistic"]): 

        ax = axes[i]
        ax.text(
            -0.18,
            1.08,
            chr(97 + i),
            transform=ax.transAxes,
            fontsize=9,
            fontweight="normal",
            va="top",
            ha="right"
        )
        
        
        sub = df_rescue[df_rescue["Target"] == tgt]

        if sub.empty:
            continue

        sub_pair = (
            sub.groupby("DomainPair", as_index=False)
            .agg({
                "IIT_score_pair": "first",
                "TransferGain": "mean"
            })
        )

        r_s, p_val, n_perm = spearman_permutation_test(
            sub_pair["IIT_score_pair"],
            sub_pair["TransferGain"]
        )

        slope, intercept, _, _, _ = linregress(
            sub_pair["IIT_score_pair"],
            sub_pair["TransferGain"]
        )

        x_range = np.linspace(
            sub_pair["IIT_score_pair"].min() - 0.2,
            sub_pair["IIT_score_pair"].max() + 0.2,
            100
        )
        
        ax.plot(x_range, intercept + slope * x_range, color='black', lw=1.2, linestyle='--', alpha=0.5)

        for dp in sub["DomainPair"].unique():
            pair_data = sub[sub["DomainPair"] == dp]; src = dp.split(" → ")[0]
            ax.scatter(
                pair_data["IIT_score_pair"],
                pair_data["TransferGain"],
                color=SOURCE_COLORS.get(src),
                s=30,
                alpha=0.45,
                edgecolors="none")
            ax.scatter(pair_data["IIT_score_pair"].mean(), pair_data["TransferGain"].mean(), 
                       color=SOURCE_COLORS.get(src), s=80, edgecolors='black', lw=1.5, label=f"From {src}")

        ax.set_title(
            rf"$r_s$ = {r_s:.2f}, "
            rf"$p_{{perm}}$ = {p_val:.3f}, "
            rf"$n$ = {len(sub_pair)}",
            fontsize=8
        )

        ax.set_xlabel(
            r"IIT score",
            fontsize=8,
            fontweight="normal"
        )

        ax.set_ylabel(
            r"Transfer gain ($\Delta$F1)",
            fontsize=8,
            fontweight="normal"
        )

        ax.legend(frameon=True, fontsize=9)
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)
        ax.tick_params(axis="both", which="major", labelsize=8, width=0.8)
        print(
            f"Target {tgt}: n_pairs = {len(sub_pair)}, "
            f"r_s = {r_s:.3f} | "
            f"permutation p = {p_val:.3f} | "
            f"permutations = {n_perm}"
        )

    save_highres(filename); plt.show()

# --- FINAL EXECUTION ---

# --- PERMUTATION TEST FOR SPEARMAN ASSOCIATION ---

SPEARMAN_PERMUTATION_SEED = 24
N_SPEARMAN_PERMUTATIONS = 100000


def spearman_permutation_test(
    x,
    y,
    n_permutations=N_SPEARMAN_PERMUTATIONS,
    seed=SPEARMAN_PERMUTATION_SEED
):
    """
    Permutation test for Spearman's rank correlation.

    The observed correlation is computed from the original paired
    observations. Under the null hypothesis, the y-values are randomly
    permuted while the x-values remain fixed.

    For n <= 8, all permutations are enumerated exactly.
    For larger n, a fixed-seed Monte Carlo permutation test is used.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    valid = np.isfinite(x) & np.isfinite(y)
    x = x[valid]
    y = y[valid]

    n = len(x)

    if n < 3:
        return np.nan, np.nan, 0

    rs_obs = spearmanr(x, y).statistic

    rng = np.random.default_rng(seed)

    if n <= 8:
        perm_iter = permutations(y)
        perm_stats = np.array([
            spearmanr(x, np.asarray(y_perm)).statistic
            for y_perm in perm_iter
        ])

        p_value = np.mean(
            np.abs(perm_stats) >=
            (np.abs(rs_obs) - 1e-12)
        )

        n_tested = len(perm_stats)

    else:
        perm_stats = np.empty(n_permutations)

        for i in range(n_permutations):
            y_perm = rng.permutation(y)

            perm_stats[i] = spearmanr(
                x,
                y_perm
            ).statistic

        # Add-one correction avoids zero Monte Carlo p-values.
        p_value = (
            1 +
            np.sum(
                np.abs(perm_stats) >=
                (np.abs(rs_obs) - 1e-12)
            )
        ) / (n_permutations + 1)

        n_tested = n_permutations

    return rs_obs, p_value, n_tested

plot_robust_quadrants(df_final, f"quadrant_validation_robust_{VALIDATION_PARADIGM}")
plot_intra_target_validation(df_final, f"intra_target_validation_{VALIDATION_PARADIGM}")

print(f"\n✅ MODULE 4 Completed. All theoretical and empirical indices synchronized.")

# ------------------------------------------------------------
# === PAIR-SPECIFIC EXACT SIGN-PERMUTATION TEST ===
# ------------------------------------------------------------
def exact_sign_permutation_test(values):
    """
    Exact two-sided sign-permutation test for a pair-specific
    TransferGain against the null hypothesis of zero effect.

    Each value corresponds to one independent experimental seed.
    For n seeds, all 2^n possible sign assignments are enumerated.
    """

    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    n = len(values)

    if n == 0:
        return np.nan, np.nan, 0

    observed_mean = np.mean(values)

    # Enumerate all possible sign assignments.
    signs = np.array(
        list(itertools.product([-1, 1], repeat=n)),
        dtype=float
    )

    perm_means = np.mean(
        signs * values[None, :],
        axis=1
    )

    p_value = np.mean(
        np.abs(perm_means) >=
        (np.abs(observed_mean) - 1e-12)
    )

    return observed_mean, p_value, len(perm_means)

# ------------------------------------------------------------
# === PAIR-SPECIFIC TRANSFER-GAIN ANALYSIS ===
# ------------------------------------------------------------
def analyze_pairwise_transfer_gains(df, filename):

    results = []

    for domain_pair, sub in df.groupby("DomainPair"):

        values = (
            sub.groupby("Seed")["TransferGain"]
            .mean()
            .dropna()
            .to_numpy()
        )

        n_seeds = len(values)

        if n_seeds == 0:
            continue

        mean_gain = np.mean(values)

        if n_seeds > 1:
            sd_gain = np.std(values, ddof=1)
            se_gain = sd_gain / np.sqrt(n_seeds)
            ci95_low = mean_gain - 1.96 * se_gain
            ci95_high = mean_gain + 1.96 * se_gain
        else:
            sd_gain = np.nan
            ci95_low = np.nan
            ci95_high = np.nan

        mean_test, p_perm, n_permutations = (
            exact_sign_permutation_test(values)
        )

        positive_fraction = np.mean(values > 0)

        iit_score = (
            sub["IIT_score_pair"]
            .dropna()
            .iloc[0]
        )

        results.append({
            "DomainPair": domain_pair,
            "IIT_score": iit_score,
            "Mean_TransferGain": mean_gain,
            "SD_TransferGain": sd_gain,
            "CI95_low": ci95_low,
            "CI95_high": ci95_high,
            "N_seeds": n_seeds,
            "Positive_seed_fraction": positive_fraction,
            "Permutation_p": p_perm,
            "N_permutations": n_permutations
        })

    df_pairwise = pd.DataFrame(results)

    # Benjamini–Hochberg correction across the 12 directed pairs.
    reject, q_values, _, _ = multipletests(
        df_pairwise["Permutation_p"],
        alpha=0.05,
        method="fdr_bh"
    )

    df_pairwise["FDR_q"] = q_values
    df_pairwise["FDR_significant"] = reject

    df_pairwise = df_pairwise.sort_values(
        "Mean_TransferGain",
        ascending=False
    ).reset_index(drop=True)

    df_pairwise.to_csv(
        filename,
        index=False
    )

    print("\n🧪 PAIR-SPECIFIC TRANSFER-GAIN ANALYSIS")
    print(
        df_pairwise[
            [
                "DomainPair",
                "IIT_score",
                "Mean_TransferGain",
                "SD_TransferGain",
                "CI95_low",
                "CI95_high",
                "N_seeds",
                "Positive_seed_fraction",
                "Permutation_p",
                "FDR_q"
            ]
        ].to_string(index=False)
    )

    print(
        f"\n📄 Pair-specific results saved to: {filename}"
    )

    return df_pairwise

# ------------------------------------------------------------
# === PAIR-SPECIFIC ANALYSIS: COMBINED STRESS ===
# ------------------------------------------------------------

df_combined_pairwise = df_final[
    (df_final["low_data_frac"] == 0.1) &
    (df_final["noise_level"] == 0.9)
].copy()

pairwise_output = os.path.join(
    results_dir,
    f"pairwise_transfer_gain_combined_{VALIDATION_PARADIGM}.csv"
)

df_pairwise_results = analyze_pairwise_transfer_gains(
    df_combined_pairwise,
    pairwise_output
)
```

```python
# ==============================================================================
# === X-CDTL Framework | Module 5: Statistical Synthesis & Stress Analysis ===
# ==============================================================================
# Researcher: Daniele Caligiore
# Affiliation: CNR - Italian National Research Council (ISTC)
# Version: 2.0 (Sep 2026)
# ------------------------------------------------------------------------------
# DESCRIPTION:
# This module provides the final statistical synthesis of the cross-domain
# transfer experiments generated in Module 3.
#
# It summarizes performance across domain pairs, random seeds, validation
# scenarios, and four predefined stress regimes. The analysis reports
# descriptive statistics, uncertainty across seeds, non-parametric comparisons,
# trajectory plots, and an integrated summary of transfer performance.
#
# The module is intended to characterize the observed transfer behaviour
# under controlled stress conditions. It does not by itself establish
# causal effects, universal transferability, or a general predictive law.
#
# CORE PIPELINE:
# 1. Synchronization and validation of Module 3 results.
# 2. Aggregation of performance metrics across seeds, domain pairs,
#    scenarios, and stress regimes.
# 3. Non-parametric comparison of the matched experimental conditions.
# 4. Visualization of performance trajectories under increasing stress.
# 5. Generation of a seed-aware summary table for the combined-stress regime.
# ------------------------------------------------------------------------------

print("\n📊 Executing X-CDTL Module 5 — Statistical Synthesis")

import os
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from scipy.stats import friedmanchisquare, wilcoxon, t

# --- NATURE COMMUNICATIONS FIGURE STYLE ---
plt.rcParams.update({
    "font.family": "sans-serif",
    "font.size": 8,
    "axes.labelsize": 8,
    "axes.titlesize": 8,
    "xtick.labelsize": 7,
    "ytick.labelsize": 7,
    "legend.fontsize": 7,
    "axes.linewidth": 0.8,
    "xtick.major.width": 0.8,
    "ytick.major.width": 0.8,
    "pdf.fonttype": 42,
    "ps.fonttype": 42
})

# --- 0. DIRECTORY & UTILITY SETUP ---
output_dir = "figures_nature"
results_dir = "results_nature"
os.makedirs(output_dir, exist_ok=True)
os.makedirs(results_dir, exist_ok=True)

def save_highres(filename_base):
    """Export current figure with extra-tight bounding box to prevent clipping."""
    plt.savefig(os.path.join(output_dir, f"{filename_base}.png"), dpi=300, bbox_inches="tight")
    plt.savefig(os.path.join(output_dir, f"{filename_base}.pdf"), bbox_inches="tight", pad_inches=0.02)
    print(f"📸 Figure saved: {filename_base}.png / .pdf")

# ------------------------------------------------------------
# === 1. INPUT VALIDATION & ABLATION SETUP ===
# ------------------------------------------------------------
if 'VALIDATION_PARADIGM' not in globals():
    VALIDATION_PARADIGM = "functional_validation"

TGI_METRIC = "f1" if VALIDATION_PARADIGM == "functional_validation" else "roc_auc"

if 'df_aligned' in globals():
    df_plot = df_aligned.copy()
else:
    raise RuntimeError("❌ df_aligned not found. Please run Module 3 first.")

# Definitions for sorting and metrics
metrics_perf = ["roc_auc", "ap", "f1"]
scenarios_order = ["No Transfer (Aligned)", "No Transfer (Augmented)", "Transfer (Aligned)"]
domain_order = ["Social", "Molecular", "Proteins", "Linguistic"]
domain_priority = {d: i for i, d in enumerate(domain_order)}

def get_pair_sort_key(pair_name):
    src = str(pair_name).split(" → ")[0]
    return domain_priority.get(src, 99)

def map_stress_ablation(row):
    a, n = row['low_data_frac'], row['noise_level']
    if a == 0.9 and n == 0.1: return "1. Baseline"
    if a == 0.1 and n == 0.1: return "2. Scarcity"
    if a == 0.9 and n == 0.9: return "3. Noise"
    if a == 0.1 and n == 0.9: return "4. Combined"
    return "Other"

df_plot['StressLevel'] = df_plot.apply(map_stress_ablation, axis=1)
df_plot = df_plot[df_plot['StressLevel'] != "Other"].copy()

# Force logical order for the stress trajectory
stress_order = ["1. Baseline", "2. Scarcity", "3. Noise", "4. Combined"]
df_plot["StressLevel"] = pd.Categorical(df_plot["StressLevel"], categories=stress_order, ordered=True)
df_plot["Scenario"] = pd.Categorical(df_plot["Scenario"], categories=scenarios_order, ordered=True)
df_plot["Target"] = df_plot["DomainPair"].apply(lambda x: str(x).split(" → ")[1])

print(f"✅ Module 5 synced. Paradigm: {VALIDATION_PARADIGM.upper()} | Observations: {len(df_plot)}")

# ------------------------------------------------------------
# === 2. AGGREGATED TRAJECTORY SUMMARY (μ ± σ) ===
# ------------------------------------------------------------
print(f"📊 Computing Statistics across Trajectory (N={len(target_seeds)} seeds)...")
global_stats = df_plot.groupby(["StressLevel", "Scenario"])[metrics_perf].agg(["mean", "std"]).round(3)
display(global_stats)

summary_detailed = df_plot.groupby(["DomainPair", "StressLevel", "Scenario"])[metrics_perf].agg(["mean", "std"]).reset_index()
for m in metrics_perf:
    summary_detailed[f"{m.upper()} (μ ± σ)"] = summary_detailed.apply(
        lambda r: f"{r[m]['mean']:.3f} ± {r[m]['std']:.3f}", axis=1)
summary_detailed.to_csv(os.path.join(results_dir, f"full_stats_{VALIDATION_PARADIGM}.csv"))

# ------------------------------------------------------------
# === 3. INFERENTIAL STATISTICS (MATCHED-DESIGN ANALYSIS) ===
# ------------------------------------------------------------
# The three scenarios are evaluated on the same DomainPair × Seed ×
# StressLevel blocks. Because the observations are therefore matched,
# a Friedman test is used for the global comparison of the three scenarios.
# Pairwise comparisons are performed with the Wilcoxon signed-rank test
# and Bonferroni correction within each metric and stress regime.
# ------------------------------------------------------------

print("\n🧠 Executing matched non-parametric inference "
      "(Friedman + Wilcoxon signed-rank)...")

inferential_results = []

for metric in metrics_perf:

    print(f"\n=== {metric.upper()} GLOBAL SIGNIFICANCE ===")

    for stress in stress_order:

        sub = df_plot[
            df_plot["StressLevel"] == stress
        ].copy()

        wide = (
            sub.pivot_table(
                index=["DomainPair", "Seed"],
                columns="Scenario",
                values=metric,
                aggfunc="mean"
            )
            .dropna(subset=scenarios_order)
        )

        if len(wide) < 2:
            print(
                f"{stress}: insufficient matched observations "
                f"for {metric.upper()}."
            )
            continue

        stat, p = friedmanchisquare(
            *[
                wide[scenario].to_numpy()
                for scenario in scenarios_order
            ]
        )

        print(
            f"{stress}: Friedman chi2={stat:.3f}, p={p:.5e}, "
            f"n_blocks={len(wide)}"
        )

        inferential_results.append({
            "Metric": metric,
            "StressLevel": stress,
            "Test": "Friedman",
            "Comparison": "All scenarios",
            "Statistic": stat,
            "p_raw": p,
            "n": len(wide)
        })

        if p < 0.05:

            pairwise_comparisons = [
                (
                    scenarios_order[0],
                    scenarios_order[1]
                ),
                (
                    scenarios_order[0],
                    scenarios_order[2]
                ),
                (
                    scenarios_order[1],
                    scenarios_order[2]
                )
            ]

            for scenario_a, scenario_b in pairwise_comparisons:

                x = wide[scenario_a].to_numpy()
                y = wide[scenario_b].to_numpy()

                # Wilcoxon is undefined when all paired differences are zero.
                if np.all(x == y):
                    stat_w = 0.0
                    p_w = 1.0
                else:
                    stat_w, p_w = wilcoxon(
                        x,
                        y,
                        alternative="two-sided"
                    )

                p_adj = min(
                    p_w * len(pairwise_comparisons),
                    1.0
                )
                # p_adjusted is populated only for pairwise Wilcoxon comparisons.
                # The omnibus Friedman test is reported using its unadjusted p-value.
                inferential_results.append({
                    "Metric": metric,
                    "StressLevel": stress,
                    "Test": "Wilcoxon signed-rank",
                    "Comparison": (
                        f"{scenario_a} vs {scenario_b}"
                    ),
                    "Statistic": stat_w,
                    "p_raw": p_w,
                    "p_adjusted": p_adj,
                    "n": len(wide)
                })

                print(
                    f"   {scenario_a} vs {scenario_b}: "
                    f"p_adj={p_adj:.5e}"
                )

df_inferential = pd.DataFrame(
    inferential_results
)

display(df_inferential)

# ------------------------------------------------------------
# === 4. BOXPLOT ANALYSIS (ROC & F1) ===
# ------------------------------------------------------------
print("\n📊 Generating Trajectory Boxplots...")
for m in ["roc_auc", "f1"]:
    plt.figure(figsize=(14, 8))
    # Comparison of all 3 scenarios (Aligned, Augmented, Transfer)
    ax = sns.boxplot(data=df_plot, x="StressLevel", y=m, hue="Scenario", 
                     palette=["#BDBDBD", "#636363", "#0072B2"], width=0.7, fliersize=0)
    sns.stripplot(data=df_plot, x="StressLevel", y=m, hue="Scenario", dodge=True, color="black", alpha=0.1, size=2, legend=False)
    plt.title(f"{m.upper()} Performance under increasing stress", fontsize=16, fontweight="bold") 
    ax.legend(
        title=None,
        frameon=False,
        fontsize=7,
        loc="best")
    sns.despine(trim=True); save_highres(f"ablation_boxplot_{m}")
    ax.set_xticklabels(["Baseline", "Scarcity", "Noise", "Combined"])
    plt.show()

# ------------------------------------------------------------
# === 5. TRAJECTORY DECAY CURVES: TRANSFER PERFORMANCE UNDER STRESS ===
# ------------------------------------------------------------
# Objective: Visualize the 'Rescue Effect' and the 'Size Control' 
# by comparing the decay of three modeling paradigms.
# ------------------------------------------------------------
print("\n📈 Plotting System Decay Curves per Target Domain (Final Robust Version)...")

import gc
plt.close('all')
gc.collect()

# 1. Setup categorical order
stress_order = ["1. Baseline", "2. Scarcity", "3. Noise", "4. Combined"]
df_plot['StressLevel'] = pd.Categorical(df_plot['StressLevel'], categories=stress_order, ordered=True)

# 2. COLOR PALETTE DEFINITION (Nature-Grade Contrast)
# We include all 3 scenarios to provide a complete scientific picture.
color_map = {
    "No Transfer (Aligned)": "#BDBDBD",    
    "No Transfer (Augmented)": "#636363",  
    "Transfer (Aligned)": "#0072B2"        
}

# Line styles to distinguish the 'Size Control' (Augmented) from the 'Structural Signal' (Transfer)
line_styles = {
    "No Transfer (Aligned)": (None, None),   # Solid
    "No Transfer (Augmented)": (2, 2),        # Dotted/Dashed
    "Transfer (Aligned)": (None, None)        # Solid
}

fig, axes = plt.subplots(
    2, 2,
    figsize=(7.0, 5.8),
    constrained_layout=False
)
axes = axes.flatten()
targets = ["Molecular", "Proteins", "Social", "Linguistic"]

for i, tgt in enumerate(targets):
    ax = axes[i]
    ax.text(
        -0.12, 1.05,
        chr(97 + i),
        transform=ax.transAxes,
        fontsize=10,
        fontweight="bold",
        va="top",
        ha="center")
    
    # Filter for target domain
    sub = df_plot[df_plot["Target"] == tgt].copy()
    
    if sub.empty: continue

    # FIX: Remove unused categories to prevent Seaborn palette errors
    sub["Scenario"] = sub["Scenario"].cat.remove_unused_categories()

    # ------------------------------------------------------------
    # Aggregate first within each seed, then across seeds.
    # Each seed contributes equally to the final estimate.
    # ------------------------------------------------------------
    seed_data = (
        sub.groupby(
            ["Seed", "StressLevel", "Scenario"],
            observed=True
        )[TGI_METRIC]
        .mean()
        .reset_index(name="seed_mean")
    )

    traj_data = (
        seed_data
        .groupby(
            ["StressLevel", "Scenario"],
            observed=True
        )["seed_mean"]
        .agg(
            mean="mean",
            std="std",
            count="count"
        )
        .reset_index()
    )

    # Standard error
    traj_data["sem"] = (
        traj_data["std"] / np.sqrt(traj_data["count"])
    )

    # 95% confidence interval across seeds
    traj_data["ci95"] = (
        t.ppf(
            0.975,
            traj_data["count"] - 1
        ) * traj_data["sem"]
    )
    
    traj_data["lower"] = (
        traj_data["mean"] - traj_data["ci95"]
    )

    traj_data["upper"] = (
        traj_data["mean"] + traj_data["ci95"]
    )

    # ------------------------------------------------------------
    # Plot mean trajectory with 95% CI
    # ------------------------------------------------------------
    x_positions = np.arange(len(stress_order))

    for scenario in scenarios_order:
        sdata = traj_data[
            traj_data["Scenario"] == scenario
            ].copy()
        if sdata.empty:
            continue

        # Ensure the stress levels follow the predefined order 
        sdata["StressLevel"] = pd.Categorical(
            sdata["StressLevel"],
            categories=stress_order,
            ordered=True
        )

        sdata = sdata.sort_values("StressLevel")
        x = np.array([
            stress_order.index(level)
            for level in sdata["StressLevel"].astype(str)
            ])
        
        y = sdata["mean"].to_numpy()
        lower = sdata["lower"].to_numpy()
        upper = sdata["upper"].to_numpy()

        ax.plot(
            x,
            y,
            color=color_map[scenario],
            linestyle="--" if scenario == "No Transfer (Augmented)" else "-",
            marker="o",
            markersize=4,
            linewidth=1.5,
            label=scenario
            )
        
        ax.fill_between(
            x,
            lower,
            upper,
            color=color_map[scenario],
            alpha=0.08,
            linewidth=0
            )
    
    # Subplot Titles
    ax.set_title(
        f"Target: {tgt}",
        fontsize=8,
        fontweight="normal",
        pad=4
    )
    
    ax.set_ylim(-0.05, 0.75)
        
    if TGI_METRIC == "f1":
        ylabel = "F1 score"
    else:
        ylabel = "ROC-AUC"

    ax.set_ylabel(
        ylabel,
        fontsize=8,
        fontweight="normal"
    )
    ax.set_xlabel("")
    
    # X-Ticks and Grids
    ax.set_xticks(range(len(stress_order)))
    display_labels = ["Baseline", "Scarcity", "Noise", "Combined"]

    ax.set_xticklabels(
        display_labels,
        rotation=0,
        ha="center",
        fontsize=7
    )

    ax.yaxis.grid(True, linestyle='--', alpha=0.3)
    sns.despine(ax=ax, trim=False)

# 4. UNIFIED GLOBAL LEGEND (With safe top placement)
# We recreate the handles to ensure the legend is perfectly formatted
from matplotlib.lines import Line2D
custom_lines = [
    Line2D(
        [0], [0],
        color=color_map["No Transfer (Aligned)"],
        lw=1.5,
        label="No transfer"
    ),
    Line2D(
        [0], [0],
        color=color_map["No Transfer (Augmented)"],
        lw=1.5,
        ls="--",
        label="No transfer + augmentation"
    ),
    Line2D(
        [0], [0],
        color=color_map["Transfer (Aligned)"],
        lw=1.5,
        label="X-CDTL transfer"
    )
]
plt.subplots_adjust(
    left=0.10,
    right=0.98,
    bottom=0.18,
    top=0.95,
    wspace=0.25,
    hspace=0.30
)
fig.legend(
    handles=custom_lines,
    loc="lower center",
    bbox_to_anchor=(0.5, 0.06),
    ncol=3,
    fontsize=7,
    frameon=False
)

# 6. EXPORT
filename = f"decay_curves_ablation_{VALIDATION_PARADIGM}"
save_highres(filename)

print(f"📸 Strategic visualization saved in {output_dir}/")
plt.show()
plt.close(fig)
gc.collect()

# ------------------------------------------------------------
# === 6. FINAL INTEGRATED MAPPING TABLE (EXECUTIVE SUMMARY) ===
# ------------------------------------------------------------
print("\n📊 Generating Comprehensive Executive Mapping Table...")

# Filter for Combined Stress (Scenario 4)
df_combined = df_plot[
    df_plot["StressLevel"] == "4. Combined"
].copy()

# ------------------------------------------------------------
# Seed-aware aggregation:
# 1. Compute the mean within each seed for each
#    DomainPair × Scenario.
# 2. Compute mean and SD across the 10 seed-level means.
# ------------------------------------------------------------
df_seed_level = (
    df_combined
    .groupby(
        ["DomainPair", "Scenario", "Seed"],
        observed=True
    )[metrics_perf]
    .mean()
    .reset_index()
)

df_final = (
    df_seed_level
    .groupby(
        ["DomainPair", "Scenario"],
        observed=True
    )[metrics_perf]
    .agg(["mean", "std"])
    .reset_index()
)

nt_df = df_final[
    df_final["Scenario"] == "No Transfer (Aligned)"
].set_index("DomainPair")

tr_df = df_final[
    df_final["Scenario"] == "Transfer (Aligned)"
].set_index("DomainPair")

mapping_table = pd.DataFrame(index=nt_df.index)

mapping_table["ROC_Baseline_mean"] = (
    nt_df[("roc_auc", "mean")]
)

mapping_table["ROC_Baseline_std"] = (
    nt_df[("roc_auc", "std")]
)

mapping_table["ROC_Transfer_mean"] = (
    tr_df[("roc_auc", "mean")]
)

mapping_table["ROC_Transfer_std"] = (
    tr_df[("roc_auc", "std")]
)

mapping_table["F1_Baseline_mean"] = (
    nt_df[("f1", "mean")]
)

mapping_table["F1_Baseline_std"] = (
    nt_df[("f1", "std")]
)

mapping_table["F1_Transfer_mean"] = (
    tr_df[("f1", "mean")]
)

mapping_table["F1_Transfer_std"] = (
    tr_df[("f1", "std")]
)

mapping_table["Abs_Gain_F1"] = (
    mapping_table["F1_Transfer_mean"]
    - mapping_table["F1_Baseline_mean"]
)

mapping_table["ROC_Baseline"] = (
    mapping_table["ROC_Baseline_mean"].map("{:.3f}".format)
    + " ± "
    + mapping_table["ROC_Baseline_std"].map("{:.3f}".format)
)

mapping_table["ROC_Transfer"] = (
    mapping_table["ROC_Transfer_mean"].map("{:.3f}".format)
    + " ± "
    + mapping_table["ROC_Transfer_std"].map("{:.3f}".format)
)

mapping_table["F1_Baseline"] = (
    mapping_table["F1_Baseline_mean"].map("{:.3f}".format)
    + " ± "
    + mapping_table["F1_Baseline_std"].map("{:.3f}".format)
)

mapping_table["F1_Transfer"] = (
    mapping_table["F1_Transfer_mean"].map("{:.3f}".format)
    + " ± "
    + mapping_table["F1_Transfer_std"].map("{:.3f}".format)
)

mapping_table["sort_key"] = [get_pair_sort_key(x) for x in mapping_table.index]

mapping_table = (
    mapping_table
    .sort_values("sort_key")
    .drop(columns=["sort_key"])
)

mapping_table_paper = mapping_table[
    [
        "ROC_Baseline",
        "ROC_Transfer",
        "F1_Baseline",
        "F1_Transfer",
        "Abs_Gain_F1"
    ]
].copy()

mapping_table_paper["Abs_Gain_F1"] = (
    mapping_table_paper["Abs_Gain_F1"]
    .round(3)
)

display(mapping_table_paper)

mapping_table_paper.to_csv(
    os.path.join(
        results_dir,
        f"executive_summary_{VALIDATION_PARADIGM}_seed_aware.csv"
    )
)

mapping_table.to_csv(
    os.path.join(
        results_dir,
        f"executive_summary_{VALIDATION_PARADIGM}_full_seed_aware.csv"
    )
)

print(f"\n🌍 EXECUTIVE SUMMARY: Critical Rescue (Scenario 4)")
display(
    mapping_table.style
    .format({
        "ROC_Baseline_mean": "{:.3f}",
        "ROC_Baseline_std": "{:.3f}",
        "ROC_Transfer_mean": "{:.3f}",
        "ROC_Transfer_std": "{:.3f}",
        "F1_Baseline_mean": "{:.3f}",
        "F1_Baseline_std": "{:.3f}",
        "F1_Transfer_mean": "{:.3f}",
        "F1_Transfer_std": "{:.3f}",
        "Abs_Gain_F1": "{:.3f}",
        "Rel_Gain_F1 (%)": "{:.1f}"
    })
)
mapping_table.to_csv(os.path.join(results_dir, f"executive_summary_{VALIDATION_PARADIGM}.csv"))

print("\n✅ MODULE 5 completed. Statistical synthesis and summary tables generated.")
```
